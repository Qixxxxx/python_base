```python
# 数值类型互转
print(int(3.14))           # 3 —— 向零截断（与 Java (int)3.14 一致）
print(int(-3.99))          # -3 —— 不是四舍五入！

print(float(42))           # 42.0
print(float("inf"))        # inf —— 支持无穷大
print(float("nan"))        # nan —— 支持非数字

# 支持进制转换（Java 的 Integer.parseInt(str, radix)）
print(int("1010", 2))      # 10 —— 二进制
print(int("FF", 16))       # 255 —— 十六进制
print(int("77", 8))        # 63 —— 八进制
```

```python
# 字符串构造 —— 所有对象都能转字符串
print(str(42))             # "42"
print(str([1, 2, 3]))      # "[1, 2, 3]" —— 调用 __str__
print(str(None))           # "None"

# repr() 与 str() 的区别：repr 面向开发者，str 面向用户
s = "hello\nworld"
print(str(s))              # hello（换行）world
print(repr(s))             # 'hello\nworld' —— 带转义，可直接复制粘贴执行
```

```python
# 数值运算的隐式提升链
# int → float → complex
print(type(1 + 2))         # <class 'int'>
print(type(1 + 2.0))       # <class 'float'> —— int 提升为 float
print(type(1.0 + 2j))      # <class 'complex'> —— float 提升为 complex

# 除法 / 总是返回 float（与 Java 的 3/2=1 完全不同！）
print(3 / 2)               # 1.5
print(type(3 / 2))         # <class 'float'>

# 地板除 // 保留整数类型（如果操作数都是 int）
print(3 // 2)              # 1
print(type(3 // 2))        # <class 'int'>
print(3.0 // 2)            # 1.0 —— 有一个 float 就返回 float
```

```python
# 序列类型互转
print(list("abc"))         # ['a', 'b', 'c'] —— 字符串拆成字符
print(list((1, 2, 3)))     # [1, 2, 3]
print(tuple([1, 2, 3]))    # (1, 2, 3)
print(set([1, 2, 2, 3]))   # {1, 2, 3} —— 去重

# range 转 list 才会展开
print(list(range(3)))      # [0, 1, 2]
```

```python
# 字典的构造艺术 —— 多种数据源转 dict
# 从二元组列表
print(dict([("a", 1), ("b", 2)]))        # {'a': 1, 'b': 2}

# 从关键字参数（仅限字符串键）
print(dict(a=1, b=2))                    # {'a': 1, 'b': 2}

# 从两个可迭代对象（zip）
keys = ["x", "y", "z"]
values = [10, 20, 30]
print(dict(zip(keys, values)))           # {'x': 10, 'y': 20, 'z': 30}

# 从字典视图回退（无意义但合法）
d = {"a": 1, "b": 2}
print(dict(d.items()))                   # {'a': 1, 'b': 2}
```

```python
# 字典与列表的局部转换
d = {"a": 1, "b": 2, "c": 3}

print(list(d))             # ['a', 'b', 'c'] —— 默认迭代 keys
print(list(d.keys()))      # 同上
print(list(d.values()))    # [1, 2, 3]
print(list(d.items()))     # [('a', 1), ('b', 2), ('c', 3)]

# 反向构造：items 转 dict
print(dict(d.items()))     # 原样还原
```

```python
# 文本与字节的互转（编码解码）
s = "中文"
b = s.encode("utf-8")        # str → bytes
print(b)                     # b'\xe4\xb8\xad\xe6\x96\x87'
print(b.decode("utf-8"))     # 中文 —— bytes → str

# 通过构造函数转换
print(bytes("abc", "ascii")) # b'abc'
print(str(b"abc", "ascii"))  # abc
```
