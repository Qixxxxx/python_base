# 9. 集合 Set

集合是用来保存不重复元素的容器，类型名是 `set`。

核心特点：
- 无序：不能通过下标访问。
- 元素唯一：重复元素会自动去重。
- 可变：可以添加和删除元素。
- 元素必须是不可变且可哈希的对象，比如数字、字符串、元组。

## 1. 集合的创建

非空集合可以用 `{}` 创建；空集合必须用 `set()` 创建，因为 `{}` 表示空字典。

```python
nums = {1, 2, 3, 2, 1}
chars = set("hello")
empty_set = set()
empty_dict = {}

print(nums)
print(chars)
print(type(empty_set))
print(type(empty_dict))
```

### 解释
- 集合会自动去重，所以 `{1, 2, 3, 2, 1}` 最终只保留 `1、2、3`。
- `set("hello")` 会遍历字符串中的字符并去重。
- 空集合必须写 `set()`，因为 `{}` 在 Python 中表示空字典。

## 2. 集合自动去重

集合常用于去重。如果需要保留原来的顺序，可以使用 `dict.fromkeys()`。

```python
nums = [1, 2, 2, 3, 3, 3, 4]

unique_nums = set(nums)
print(unique_nums)

unique_keep_order = list(dict.fromkeys(nums))
print(unique_keep_order)
```

### 解释
- `set(nums)` 可以快速去重，但集合无序，结果顺序不一定和原列表一致。
- `dict.fromkeys(nums)` 会把列表元素当作字典键，利用键唯一的特点去重。
- 从 Python 3.7 开始，普通字典会保持插入顺序，所以这种方式可以保留原顺序。

## 3. 添加元素

| 方法 | 作用 |
| --- | --- |
| `add(x)` | 添加一个元素 |
| `update(iterable)` | 添加多个元素 |

```python
s = {1, 2, 3}

s.add(4)
print(s)

s.update([4, 5, 6])
print(s)

s.update("ab")
print(s)
```

### 解释
- `add(4)` 是添加一个整体元素。
- `update([4, 5, 6])` 会遍历列表，把里面的元素逐个加入集合。
- `update("ab")` 会把字符串拆成 `'a'` 和 `'b'` 两个字符加入集合。

## 4. 删除元素

| 方法 | 作用 |
| --- | --- |
| `remove(x)` | 删除元素，元素不存在会报错 |
| `discard(x)` | 删除元素，元素不存在也不会报错 |
| `pop()` | 随机删除并返回一个元素 |
| `clear()` | 清空集合 |

```python
s = {1, 2, 3, 4}

s.remove(1)
print(s)

s.discard(99)       # 不报错
print(s)

item = s.pop()
print(item, s)

s.clear()
print(s)
```

### 解释
- `remove()` 要求元素必须存在，不存在就报错。
- `discard()` 更温和，元素不存在也什么都不做，适合“不确定有没有”的场景。
- `pop()` 删除哪个元素不应该依赖，因为集合本身无序。

## 5. 成员判断和遍历

集合适合做成员判断，`x in set` 通常比 `x in list` 更快。集合无序，所以遍历顺序不固定。

```python
allowed_users = {"Tom", "Jerry", "Alice"}

print("Tom" in allowed_users)
print("Bob" not in allowed_users)

for user in allowed_users:
    print(user)
```

### 解释
- 集合非常适合判断“某个元素是否存在”。
- 集合没有下标，不能写 `allowed_users[0]`。
- 遍历集合时不要依赖输出顺序，顺序可能和你写入时不一样。

## 6. 集合运算

| 运算 | 符号 | 方法 | 含义 |
| --- | --- | --- | --- |
| 并集 | `a | b` | `a.union(b)` | 两边所有元素 |
| 交集 | `a & b` | `a.intersection(b)` | 两边共有元素 |
| 差集 | `a - b` | `a.difference(b)` | 属于 a 但不属于 b |
| 对称差集 | `a ^ b` | `a.symmetric_difference(b)` | 只出现在其中一边的元素 |

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)
print(a & b)
print(a - b)
print(b - a)
print(a ^ b)
```

### 解释
- 并集 `|` 表示把两边的元素合在一起并去重。
- 交集 `&` 表示只保留两边都出现的元素。
- 差集 `a - b` 有方向：表示在 `a` 中但不在 `b` 中的元素。

## 7. 子集、超集和不相交

这些判断常用于权限、标签、课程选择等场景。

```python
required = {"python", "sql"}
skills = {"python", "sql", "excel"}
frontend = {"html", "css"}

print(required <= skills)          # required 是否是 skills 的子集
print(skills >= required)          # skills 是否是 required 的超集
print(skills.isdisjoint(frontend)) # 是否没有交集
```

### 解释
- `required <= skills` 判断必需技能是否都包含在已有技能中。
- `skills >= required` 是同一个关系的反向表达，表示已有技能是否覆盖必需技能。
- `isdisjoint()` 判断两个集合是否完全没有共同元素。

## 8. 集合推导式

集合推导式可以快速生成集合，并自动去重。

```python
nums = [1, 2, 2, 3, 4, 4]

squares = {x * x for x in nums}
even_squares = {x * x for x in nums if x % 2 == 0}

print(squares)
print(even_squares)
```

### 解释
- 集合推导式和列表推导式很像，只是外层使用 `{}`。
- 生成集合时会自动去重，所以相同平方结果只会出现一次。
- 如果需要保留重复结果，应该使用列表而不是集合。

## 9. 集合元素必须可哈希

列表、字典、集合是可变对象，不能作为集合元素。元组通常可以作为集合元素，但如果元组里面包含列表，也不行。

```python
ok = {(1, 2), (3, 4)}
print(ok)

# bad = {[1, 2], [3, 4]}       # TypeError：列表不可哈希
# bad_tuple = {([1, 2], 3)}    # TypeError：元组里包含列表，也不可哈希
```

### 解释
- 集合内部需要通过哈希值快速判断元素是否已存在。
- 可变对象的内容会变，哈希值不稳定，所以列表、字典、集合不能作为集合元素。
- 元组本身不可变，但如果内部包含列表，整体仍然不可哈希。

## 10. `frozenset`

`frozenset` 是不可变集合。普通集合可变，不能作为字典的键；`frozenset` 不可变，可以作为字典的键。

```python
a = frozenset([1, 2, 3])
print(a)

data = {
    frozenset(["python", "sql"]): "数据分析基础"
}
print(data[frozenset(["sql", "python"])])
```

### 解释
- `frozenset` 创建后不能再添加或删除元素。
- 因为它不可变，所以可以作为字典键，也可以作为另一个集合的元素。
- 当你想表达“一组固定标签”并拿它当键时，`frozenset` 很有用。

## 11. 集合和列表的选择

| 类型 | 特点 | 适合场景 |
| --- | --- | --- |
| `list` | 有序、可重复、可下标访问 | 保存一串有顺序的数据 |
| `set` | 无序、唯一、成员判断快 | 去重、交并差运算、快速判断存在性 |
