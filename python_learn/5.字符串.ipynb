# 5. 字符串 String

字符串是 Python 中用来表示文本的数据类型，类型名是 `str`。

核心特点：
- 有序：字符串中的每个字符都有位置，可以用下标访问。
- 不可变：字符串创建后不能原地修改，只能生成新的字符串。
- 可迭代：可以用 `for` 循环逐个取出字符。
- 支持切片：可以快速截取一段文本。

## 1. 字符串的创建

可以使用单引号、双引号、三引号创建字符串。单引号和双引号没有本质区别，三引号常用于多行字符串。

```python
s1 = 'hello'
s2 = "Python"
s3 = """第一行
第二行
第三行"""

print(s1)
print(s2)
print(s3)
print(type(s1))
```

### 解释
- `s1` 和 `s2` 都是普通字符串，单引号和双引号只是写法不同，运行效果一样。
- 当字符串内容本身包含单引号时，可以外层用双引号；包含双引号时，可以外层用单引号，这样能少写转义符。
- 三引号字符串会保留换行，适合写多行文本、SQL、模板文本等。

## 2. 转义字符和原始字符串

反斜杠 `\` 可以让某些字符拥有特殊含义，比如换行 `\n`、制表符 `\t`。如果不想让反斜杠转义，可以在字符串前面加 `r`，得到原始字符串。

```python
print("hello\nworld")     # \n 表示换行
print("name\tage")        # \t 表示制表符
print("I\'m OK")          # 在单引号字符串里表示单引号

path = r"C:\new\test.txt"
print(path)                # r 开头：反斜杠按普通字符处理
```

### 解释
- `
` 不会显示成两个字符，而是被解释成“换行”。`	` 会被解释成“制表符”。
- 如果想在字符串中显示引号，可以使用转义符，也可以调整外层引号类型。
- `r"..."` 是原始字符串，常用于 Windows 路径、正则表达式等反斜杠很多的场景。

## 3. 下标和切片

下标从 `0` 开始，负数下标从右往左数。切片语法：`字符串[开始:结束:步长]`，结束位置取不到。

```python
text = "Python"

print(text[0])       # 第 1 个字符：P
print(text[-1])      # 最后 1 个字符：n
print(text[0:2])     # 下标 0 到 2，不包含 2：Py
print(text[2:])      # 从下标 2 到最后：thon
print(text[:4])      # 从开头到下标 4，不包含 4：Pyth
print(text[::2])     # 每隔 2 个取 1 个：Pto
print(text[::-1])    # 反转字符串：nohtyP
```

### 解释
- Python 下标从 `0` 开始，所以 `text[0]` 取第一个字符。
- 切片遵循“左闭右开”：`text[0:2]` 包含下标 `0`，不包含下标 `2`。
- 步长为负数时表示从右向左取，`[::-1]` 是常见的反转写法。

## 4. 字符串不可变

不能通过下标直接修改字符串中的某个字符。看起来像修改字符串的操作，本质上都是创建了一个新字符串。

```python
word = "cat"
# word[0] = "b"       # TypeError：字符串不支持原地修改

new_word = "b" + word[1:]
print(word)           # 原字符串不变
print(new_word)       # 新字符串：bat
```

### 解释
- 字符串是不可变对象，所以 `word[0] = "b"` 这种写法会报错。
- 想“修改”字符串时，通常是取出原字符串的一部分，再和新内容拼接成另一个字符串。
- 变量 `word` 指向的原字符串没有变化，`new_word` 指向的是新创建的字符串。

## 5. 遍历、长度和成员判断

- `len()`：获取字符串长度。
- `in`：判断某个字符或子串是否在字符串中。
- `for`：逐个字符遍历。

```python
msg = "hello"

print(len(msg))
print("he" in msg)
print("x" not in msg)

for ch in msg:
    print(ch)
```

### 解释
- `len(msg)` 返回字符串长度，也就是字符个数。
- `in` 可以判断一个字符或一段子串是否存在于字符串中。
- `for ch in msg` 会按顺序取出每个字符，适合做逐字符统计、过滤、转换等操作。

## 6. 拼接和重复

`+` 可以拼接字符串，`*` 可以重复字符串。注意：字符串只能和字符串拼接，数字需要先转成字符串。

```python
name = "小明"
age = 18

print("你好，" + name)
print("ha" * 3)
print(name + "今年" + str(age) + "岁")
```

### 解释
- `+` 拼接字符串时，左右两边都必须是字符串，不能直接把数字和字符串相加。
- `str(age)` 把整数转成字符串后，才能参与字符串拼接。
- 实际开发中更推荐 f-string，比如 `f"{name}今年{age}岁"`，可读性更好。

## 7. 常用查询和替换方法

| 方法 | 作用 |
| --- | --- |
| `find()` | 查找子串位置，找不到返回 `-1` |
| `index()` | 查找子串位置，找不到会报错 |
| `count()` | 统计出现次数 |
| `replace()` | 替换内容，返回新字符串 |
| `startswith()` | 判断是否以某内容开头 |
| `endswith()` | 判断是否以某内容结尾 |

```python
sentence = "Python is easy, Python is useful"

print(sentence.find("Python"))
print(sentence.find("Java"))
print(sentence.count("Python"))
print(sentence.replace("Python", "Java", 1))  # 只替换第 1 次
print(sentence.startswith("Python"))
print(sentence.endswith("useful"))
```

### 解释
- `find()` 找不到时返回 `-1`，适合用于“先查一下有没有”的场景。
- `index()` 找不到会报错，更适合确定目标一定存在时使用。
- `replace(old, new, count)` 的第三个参数表示最多替换几次，不写时会全部替换。

## 8. 大小写和空白处理

这些方法经常用于清洗用户输入。它们不会修改原字符串，而是返回处理后的新字符串。

```python
s = "  hello Python  "

print(s.strip())       # 去掉两边空白
print(s.lstrip())      # 去掉左边空白
print(s.rstrip())      # 去掉右边空白

word = "python"
print(word.upper())
print("PYTHON".lower())
print("hello python".title())
print("hello python".capitalize())
```

### 解释
- `strip()` 只去掉字符串两端的空白，不会删除中间的空格。
- `upper()`、`lower()` 常用于统一大小写，比如验证码、用户名匹配。
- 这些方法都不会修改原字符串，而是返回处理后的新字符串。

## 9. 分割和合并

- `split()`：把字符串拆成列表。
- `join()`：把多个字符串合成一个字符串。

`join()` 的写法是 `连接符.join(字符串序列)`。

```python
data = "apple,banana,orange"
fruits = data.split(",")
print(fruits)

new_data = " | ".join(fruits)
print(new_data)

line = "  Tom   18   Beijing  "
print(line.split())    # 不传参数时，会按连续空白分割
```

### 解释
- `split(",")` 表示按逗号切开，结果是一个列表。
- `join()` 和 `split()` 相反，它把多个字符串按指定连接符合并成一个字符串。
- `split()` 不传参数时，会自动按连续空白分割，并忽略首尾空白。

## 10. 判断字符串内容

| 方法 | 判断内容 |
| --- | --- |
| `isdigit()` | 是否全是数字字符 |
| `isalpha()` | 是否全是字母或文字字符 |
| `isalnum()` | 是否全是字母、文字或数字 |
| `isspace()` | 是否全是空白字符 |

```python
print("123".isdigit())
print("abc".isalpha())
print("abc123".isalnum())
print("   \n".isspace())

user_input = "18"
if user_input.isdigit():
    age = int(user_input)
    print(age + 1)
```

### 解释
- `isdigit()` 只能判断“全是数字字符”的字符串，像 `"-1"`、`"3.14"` 都不是纯数字字符。
- 这类判断方法遇到空字符串通常返回 `False`。
- 用户输入默认是字符串，先判断是否合法，再转换类型，程序会更稳。

## 11. 字符串格式化

推荐优先使用 f-string，写法简单，可读性强。

```python
name = "小红"
score = 95.678

print(f"{name} 的成绩是 {score}")
print(f"{name} 的成绩是 {score:.2f}")     # 保留 2 位小数
print("{} 的成绩是 {:.1f}".format(name, score))
print("%s 的成绩是 %.1f" % (name, score))
```

### 解释
- f-string 在字符串前加 `f`，花括号里可以直接写变量或表达式。
- `:.2f` 表示把数字按浮点数格式显示，并保留 2 位小数。
- `%` 格式化是较旧写法，`format()` 是通用写法，初学阶段优先掌握 f-string。

## 12. 编码和解码

- `str`：文本字符串。
- `bytes`：字节数据。
- `encode()`：字符串编码成字节。
- `decode()`：字节解码成字符串。

```python
text = "中文"
data = text.encode("utf-8")

print(data)
print(type(data))
print(data.decode("utf-8"))
```

### 解释
- `encode()` 把人能读懂的文本转成计算机更底层的字节数据。
- `decode()` 把字节数据按指定编码还原成文本。
- 编码和解码要使用同一种规则，比如都用 `utf-8`，否则可能出现乱码或报错。
