# 10. 条件控制语句

条件控制语句用来让程序根据不同情况执行不同代码。简单说，就是让程序具备“判断能力”。

本节详细整理 5 类常见条件控制写法：

- 单分支 `if`
- 双分支 `if...else`
- 多分支 `if...elif...else`
- 嵌套 `if`
- `match...case`，Python 3.10 及以上支持

每种语句后面都有 20 道循序渐进练习题。

## 0. 条件判断的基础

条件控制语句后面接的是一个条件表达式。这个表达式最终会被 Python 判断成 `True` 或 `False`。

常用来构造条件的内容：

| 类型 | 示例 | 说明 |
| --- | --- | --- |
| 比较运算符 | `age >= 18` | 比较两个值的大小或是否相等 |
| 逻辑运算符 | `age >= 18 and score >= 60` | 把多个条件组合起来 |
| 成员运算符 | `name in names` | 判断元素是否存在 |
| 布尔值 | `is_login` | 直接使用布尔变量 |
| 对象真假值 | `if data:` | 空值为假，非空值通常为真 |

````python
# 条件表达式最终会得到 True 或 False
age = 20
score = 85
names = ["Tom", "Jerry", "Alice"]
is_login = True

print(age >= 18)                         # 比较判断
print(age >= 18 and score >= 60)         # 多个条件同时满足
print("Tom" in names)                    # 判断元素是否在列表中
print(is_login)                          # 直接使用布尔变量
print(bool([]))                           # 空列表是假
print(bool([1, 2, 3]))                    # 非空列表是真
````

### 解释

- `if` 后面的条件不一定要直接写 `True` 或 `False`，比较表达式、逻辑表达式都会得到布尔结果。
- `and` 表示所有条件都为真，整体才为真。
- `or` 表示只要有一个条件为真，整体就为真。
- `not` 表示取反，`not True` 是 `False`，`not False` 是 `True`。
- Python 中空字符串、空列表、空字典、空集合、数字 `0`、`None` 都会被当作假。

## 0.1 缩进非常重要

Python 不使用 `{}` 表示代码块，而是用缩进表示代码属于哪一层。条件成立后，要执行的代码必须缩进。

````python
age = 18

if age >= 18:
    print("已经成年")
    print("可以办理成年相关业务")

print("这行没有缩进，所以不属于 if 代码块")
````

### 解释

- 冒号 `:` 表示下面要开始一个代码块。
- 缩进的代码属于 `if` 语句内部，只有条件成立时才执行。
- 没有缩进的代码已经回到外层，不管条件是否成立都会继续执行。
- 建议统一使用 4 个空格缩进，不要混用 Tab 和空格。

## 1. 单分支 `if` 语句

单分支 `if` 语句只有一个条件：条件成立就执行代码，条件不成立就跳过。

语法：

```python
if 条件:
    条件成立时执行的代码
```

适合场景：只关心“满足条件时做一件事”，不需要处理“不满足”的情况。

````python
age = 20

if age >= 18:
    print("你已经成年")

print("程序继续执行")
````

### 解释

- `age >= 18` 是条件表达式。
- 条件为 `True` 时，缩进的 `print("你已经成年")` 会执行。
- 条件为 `False` 时，缩进代码会被跳过。
- 最后一行没有缩进，所以不属于 `if`，无论条件是否成立都会执行。

````python
cart = ["苹果", "牛奶"]

if cart:
    print("购物车中有商品，可以结算")
````

### 解释

- `if cart:` 利用了列表的真假值。
- 空列表 `[]` 会被判断为假，非空列表会被判断为真。
- 这种写法比 `if len(cart) > 0:` 更简洁，是 Python 中很常见的风格。

### 单分支 `if` 练习题 10 道（含参考代码）

1. 定义变量 `age = 18`，如果年龄大于等于 18，输出“已成年”。
2. 定义变量 `score = 70`，如果成绩大于等于 60，输出“及格”。
3. 定义变量 `num = 10`，如果数字大于 0，输出“正数”。
4. 定义变量 `name = "Tom"`，如果名字去掉首尾空格后不为空，输出“用户名已填写”。
5. 定义列表 `items = ["book"]`，如果列表不为空，输出“有商品”。
6. 定义变量 `temperature = 35`，如果温度大于 30，输出“天气炎热”。
7. 定义变量 `password = "abc123"`，如果密码长度大于等于 6，输出“密码长度合格”。
8. 定义变量 `filename = "demo.py"`，如果文件名以 `.py` 结尾，输出“这是 Python 文件”。
9. 定义一个列表，如果数字 `3` 在列表中，输出“找到了 3”。
10. 输入一句话，如果长度超过 20，输出“这句话比较长”。

````python
# 单分支 if：只在条件成立时执行代码

# 1. 判断是否成年
age = 18
if age >= 18:
    print("1. 已成年")

# 2. 判断是否及格
score = 70
if score >= 60:
    print("2. 及格")

# 3. 判断是否为正数
num = 10
if num > 0:
    print("3. 正数")

# 4. 判断用户名是否已填写
name = "Tom"
if name.strip():  # strip() 去掉首尾空格，非空字符串会被判断为 True
    print("4. 用户名已填写")

# 5. 判断列表是否非空
items = ["book"]
if items:  # 非空列表为 True，空列表 [] 为 False
    print("5. 有商品")

# 6. 判断天气是否炎热
temperature = 35
if temperature > 30:
    print("6. 天气炎热")

# 7. 判断密码长度是否合格
password = "abc123"
if len(password) >= 6:
    print("7. 密码长度合格")

# 8. 判断是否为 Python 文件
filename = "demo.py"
if filename.endswith(".py"):
    print("8. 这是 Python 文件")

# 9. 判断元素是否在列表中
nums = [1, 2, 3, 4, 5]
if 3 in nums:
    print("9. 找到了 3")

# 10. 判断句子是否较长
sentence = "Python 条件控制语句可以让程序根据情况执行不同代码"
if len(sentence) > 20:
    print("10. 这句话比较长")
````

## 2. 双分支 `if...else` 语句

双分支用于处理两种互斥情况：条件成立执行一段代码，条件不成立执行另一段代码。

语法：

```python
if 条件:
    条件成立时执行
else:
    条件不成立时执行
```

适合场景：事情一定会分成两类，比如及格/不及格、偶数/奇数、登录成功/登录失败。

````python
score = 55

if score >= 60:
    print("及格")
else:
    print("不及格")
````

### 解释

- `if` 和 `else` 只会执行其中一个分支。
- 条件 `score >= 60` 为真时，执行 `if` 下面的代码。
- 条件为假时，执行 `else` 下面的代码。
- `else` 后面不写条件，因为它表示“以上条件都不满足”。

````python
num = 9

if num % 2 == 0:
    print("偶数")
else:
    print("奇数")
````

### 解释

- `%` 是取余运算符。
- 一个整数除以 2 的余数为 0，说明它是偶数。
- 如果余数不为 0，就会进入 `else` 分支，被判断为奇数。

### 双分支 `if...else` 练习题 10 道（含参考代码）

1. 输入一个年龄，如果大于等于 18，输出“成年”，否则输出“未成年”。

4. 输入一个整数，判断它是正数还是非正数。

8. 输入两个数字，输出较大的那个。

````python
# 双分支 if...else：两种情况一定执行其中一种

# 1. 成年或未成年
age = 16
if age >= 18:
    print("1. 成年")
else:
    print("1. 未成年")

# 2. 及格或不及格
score = 58
if score >= 60:
    print("2. 及格")
else:
    print("2. 不及格")

# 3. 奇数或偶数
num = 9
if num % 2 == 0:
    print("3. 偶数")
else:
    print("3. 奇数")

# 4. 正数或非正数
num = -3
if num > 0:
    print("4. 正数")
else:
    print("4. 非正数")

# 5. 管理员或普通用户
username = "admin"
if username == "admin":
    print("5. 管理员")
else:
    print("5. 普通用户")

# 6. 密码安全或太短
password = "abc123"
if len(password) >= 8:
    print("6. 安全")
else:
    print("6. 太短")

# 7. 包邮或不包邮
amount = 88
if amount >= 100:
    print("7. 包邮")
else:
    print("7. 不包邮")

# 8. 输出较大的数字
a = 12
b = 20
if a >= b:
    print("8. 较大的数字是", a)
else:
    print("8. 较大的数字是", b)

# 9. 判断两次密码是否一致
password1 = "abc12345"
password2 = "abc12345"
if password1 == password2:
    print("9. 两次密码一致")
else:
    print("9. 两次密码不一致")

# 10. 判断体温是否发热
temperature = 38.1
if temperature > 37.3:
    print("10. 发热")
else:
    print("10. 正常")
````

## 3. 多分支 `if...elif...else` 语句

多分支用于处理三种及以上情况。Python 会从上到下依次判断，遇到第一个为真的条件后执行对应代码，然后跳过剩余分支。

语法：

```python
if 条件1:
    条件1成立时执行
elif 条件2:
    条件2成立时执行
elif 条件3:
    条件3成立时执行
else:
    以上条件都不成立时执行
```

适合场景：成绩等级、月份季节、菜单选择、会员等级、订单状态等。

````python
score = 86

if score >= 90:
    print("优秀")
elif score >= 80:
    print("良好")
elif score >= 60:
    print("及格")
else:
    print("不及格")
````

### 解释

- 多分支判断的顺序很重要。
- `86 >= 90` 不成立，所以继续判断下一个条件。
- `86 >= 80` 成立，输出“良好”。
- 一旦某个分支成立，后面的 `elif` 和 `else` 就不会再判断。

````python
level = "gold"

if level == "normal":
    discount = 1.0
elif level == "silver":
    discount = 0.9
elif level == "gold":
    discount = 0.8
elif level == "diamond":
    discount = 0.7
else:
    discount = 1.0

print(f"折扣为：{discount}")
````

### 解释

- 多分支可以把一个变量和多个可能的值逐个比较。
- `else` 可以作为兜底分支，处理未知等级或异常输入。
- 如果分支很多，并且只是做“值到值”的映射，后面也可以考虑用字典简化。

### 多分支 `if...elif...else` 练习题 10 道（含参考代码）

1. 输入一个成绩，输出“不及格、及格、良好、优秀”四个等级。
2. 输入一个年龄，输出“儿童、少年、青年、中年、老年”。
3. 输入一个月份，输出对应季节。
4. 输入一个星期数字 1 到 7，输出星期一到星期日，否则输出“输入错误”。
5. 输入一个会员等级 `normal/silver/gold/diamond`，输出对应折扣。
6. 输入一个订单状态 `0/1/2/3`，分别输出“待付款、待发货、运输中、已完成”。
7. 输入一个数字，判断它是正数、负数还是 0。
8. 输入一个 BMI 数值，输出“偏瘦、正常、超重、肥胖”。
9. 输入一个运算符 `+ - * /`，输出两个数字的计算结果。
10. 输入一个 HTTP 状态码，输出“成功、重定向、客户端错误、服务器错误、未知状态”。

````python
# 多分支 if...elif...else：从上到下判断，命中一个分支后就结束

# 1. 成绩等级
score = 86
if score >= 90:
    print("1. 优秀")
elif score >= 80:
    print("1. 良好")
elif score >= 60:
    print("1. 及格")
else:
    print("1. 不及格")

# 2. 年龄阶段
age = 25
if age < 6:
    print("2. 儿童")
elif age < 18:
    print("2. 少年")
elif age < 35:
    print("2. 青年")
elif age < 60:
    print("2. 中年")
else:
    print("2. 老年")

# 3. 月份对应季节
month = 7
if month in [3, 4, 5]:
    print("3. 春季")
elif month in [6, 7, 8]:
    print("3. 夏季")
elif month in [9, 10, 11]:
    print("3. 秋季")
elif month in [12, 1, 2]:
    print("3. 冬季")
else:
    print("3. 月份错误")

# 4. 星期数字
weekday = 3
if weekday == 1:
    print("4. 星期一")
elif weekday == 2:
    print("4. 星期二")
elif weekday == 3:
    print("4. 星期三")
elif weekday == 4:
    print("4. 星期四")
elif weekday == 5:
    print("4. 星期五")
elif weekday == 6:
    print("4. 星期六")
elif weekday == 7:
    print("4. 星期日")
else:
    print("4. 输入错误")

# 5. 会员折扣
level = "gold"
if level == "normal":
    discount = 1.0
elif level == "silver":
    discount = 0.9
elif level == "gold":
    discount = 0.8
elif level == "diamond":
    discount = 0.7
else:
    discount = 1.0  # 未知等级按无折扣处理
print("5. 折扣：", discount)

# 6. 订单状态
status = 2
if status == 0:
    print("6. 待付款")
elif status == 1:
    print("6. 待发货")
elif status == 2:
    print("6. 运输中")
elif status == 3:
    print("6. 已完成")
else:
    print("6. 状态错误")

# 7. 正数、负数或 0
num = 0
if num > 0:
    print("7. 正数")
elif num < 0:
    print("7. 负数")
else:
    print("7. 0")

# 8. BMI 分类
bmi = 24.5
if bmi < 18.5:
    print("8. 偏瘦")
elif bmi < 24:
    print("8. 正常")
elif bmi < 28:
    print("8. 超重")
else:
    print("8. 肥胖")

# 9. 简单计算器
a = 10
b = 2
op = "/"
if op == "+":
    result = a + b
elif op == "-":
    result = a - b
elif op == "*":
    result = a * b
elif op == "/":
    # 除法要额外考虑除数不能为 0
    if b != 0:
        result = a / b
    else:
        result = "除数不能为 0"
else:
    result = "未知运算符"
print("9. 计算结果：", result)

# 10. HTTP 状态码分类
code = 404
if 200 <= code < 300:
    print("10. 成功")
elif 300 <= code < 400:
    print("10. 重定向")
elif 400 <= code < 500:
    print("10. 客户端错误")
elif 500 <= code < 600:
    print("10. 服务器错误")
else:
    print("10. 未知状态")
````

## 4. 嵌套 `if` 语句

嵌套 `if` 是指在一个 `if` 代码块里面继续写另一个 `if`。它适合处理“先满足大条件，再继续判断小条件”的场景。

语法：

```python
if 外层条件:
    if 内层条件:
        外层和内层都成立时执行
    else:
        外层成立但内层不成立时执行
else:
    外层不成立时执行
```

适合场景：登录后再判断权限、成绩合格后再判断等级、商品有库存后再判断余额是否足够。

````python
is_login = True
role = "admin"

if is_login:
    if role == "admin":
        print("进入后台管理页面")
    else:
        print("进入普通用户页面")
else:
    print("请先登录")
````

### 解释

- 程序先判断外层条件 `is_login`。
- 只有登录成功后，才需要继续判断用户角色。
- 如果外层条件不成立，内层 `if` 根本不会执行。
- 嵌套层级不要太深，否则代码会越来越难读。

````python
stock = 5
money = 80
price = 60

if stock > 0:
    if money >= price:
        print("购买成功")
    else:
        print("余额不足")
else:
    print("商品缺货")
````

### 解释

- 先判断库存，是因为没有库存时就不需要继续判断余额。
- 只有 `stock > 0` 成立，才会进入内层判断 `money >= price`。
- 嵌套 `if` 可以让逻辑更贴近真实流程：先判断前置条件，再判断下一步条件。

### 嵌套 `if` 练习题 10 道（含参考代码）

1. 先判断用户是否登录，登录后再判断是否是管理员。
2. 先判断商品是否有库存，有库存后再判断余额是否足够。
3. 先判断成绩是否及格，及格后再判断是否优秀。
4. 先判断年龄是否成年，成年后再判断是否有驾照。
5. 先判断用户名是否正确，正确后再判断密码是否正确。
6. 先判断列表是否为空，非空后再判断第一个元素是否为正数。
7. 先判断邮箱是否包含 `@`，包含后再判断是否以 `.com` 结尾。
8. 先判断年份能否被 4 整除，再继续判断是否能被 100 和 400 整除，完成闰年判断。
9. 先判断输入是否为数字字符串，是数字后再转换成整数并判断奇偶。
10. 设计一个取款流程：先判断银行卡是否有效，再判断密码，再判断余额是否足够。

````python
# 嵌套 if：先判断外层前置条件，再判断内层细节条件

# 1. 登录后判断角色
is_login = True
role = "admin"
if is_login:
    if role == "admin":
        print("1. 管理员页面")
    else:
        print("1. 普通用户页面")
else:
    print("1. 请先登录")

# 2. 有库存后判断余额
stock = 5
money = 80
price = 60
if stock > 0:
    if money >= price:
        print("2. 购买成功")
    else:
        print("2. 余额不足")
else:
    print("2. 商品缺货")

# 3. 及格后判断是否优秀
score = 92
if score >= 60:
    if score >= 90:
        print("3. 及格，并且优秀")
    else:
        print("3. 及格，但不是优秀")
else:
    print("3. 不及格")

# 4. 成年后判断是否有驾照
age = 20
has_license = True
if age >= 18:
    if has_license:
        print("4. 可以开车")
    else:
        print("4. 成年了，但没有驾照")
else:
    print("4. 未成年，不能开车")

# 5. 用户名正确后判断密码
username = "admin"
password = "123456"
if username == "admin":
    if password == "123456":
        print("5. 登录成功")
    else:
        print("5. 密码错误")
else:
    print("5. 用户名错误")

# 6. 非空列表后判断第一个元素
nums = [3, -1, 5]
if nums:
    if nums[0] > 0:
        print("6. 第一个元素是正数")
    else:
        print("6. 第一个元素不是正数")
else:
    print("6. 列表为空")

# 7. 邮箱格式初步判断
email = "test@example.com"
if "@" in email:
    if email.endswith(".com"):
        print("7. 可能是 .com 邮箱")
    else:
        print("7. 包含 @，但不是 .com 邮箱")
else:
    print("7. 邮箱缺少 @")

# 8. 闰年判断
# 规则：能被 4 整除但不能被 100 整除，或者能被 400 整除，就是闰年
year = 2024
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("8. 闰年")
        else:
            print("8. 平年")
    else:
        print("8. 闰年")
else:
    print("8. 平年")

# 9. 先判断是否为数字字符串，再判断奇偶
text = "15"
if text.isdigit():
    num = int(text)  # 确认是数字字符串后再转换，避免转换时报错
    if num % 2 == 0:
        print("9. 偶数")
    else:
        print("9. 奇数")
else:
    print("9. 输入的不是数字")

# 10. 取款流程
card_valid = True
pin_correct = True
balance = 500
amount = 300
if card_valid:
    if pin_correct:
        if balance >= amount:
            print("10. 取款成功")
        else:
            print("10. 余额不足")
    else:
        print("10. 密码错误")
else:
    print("10. 银行卡无效")
````

## 5. `match...case` 语句

`match...case` 是 Python 3.10 新增的模式匹配语句，适合处理“一个值有很多固定可能”的情况。

语法：

```python
match 变量:
    case 模式1:
        匹配模式1时执行
    case 模式2:
        匹配模式2时执行
    case _:
        以上都不匹配时执行
```

`case _` 相当于默认分支，类似 `if...elif...else` 中的 `else`。

````python
command = "start"

match command:
    case "start":
        print("启动程序")
    case "stop":
        print("停止程序")
    case "restart":
        print("重启程序")
    case _:
        print("未知命令")
````

### 解释

- `match command:` 表示拿 `command` 的值去匹配下面的每个 `case`。
- 匹配到 `"start"` 后，执行对应分支。
- `_` 是通配符，表示其他所有情况。
- 当分支只是“某个固定值对应某个处理”时，`match...case` 比长串 `elif` 更清晰。

````python
status = 404

match status:
    case 200 | 201:
        print("请求成功")
    case 400 | 401 | 403:
        print("客户端请求错误")
    case 404:
        print("资源不存在")
    case 500:
        print("服务器错误")
    case _:
        print("未知状态码")
````

### 解释

- `case 200 | 201:` 表示匹配 200 或 201，只要满足其中一个就进入该分支。
- 多个固定值对应同一种处理时，可以用 `|` 合并。
- `match...case` 的顺序也很重要，会从上到下寻找第一个匹配项。

````python
point = (3, 0)

match point:
    case (0, 0):
        print("原点")
    case (x, 0):
        print(f"x 轴上的点，x={x}")
    case (0, y):
        print(f"y 轴上的点，y={y}")
    case (x, y):
        print(f"普通点，x={x}, y={y}")
````

### 解释

- `match...case` 不只可以匹配普通值，也可以匹配元组、列表等结构。
- `case (x, 0):` 表示第二个元素必须是 0，第一个元素可以被取出来保存到变量 `x`。
- 这种结构匹配是 `match...case` 比普通 `if...elif` 更强的地方。
- 初学时先掌握固定值匹配和 `_` 默认分支即可。

### `match...case` 练习题 10 道（含参考代码）

1. 输入一个命令 `start/stop/restart`，用 `match...case` 输出对应操作。
2. 输入星期数字 1 到 7，输出星期几。
3. 输入月份数字 1 到 12，输出所在季度。
4. 输入会员等级 `normal/silver/gold/diamond`，输出对应折扣。
5. 输入交通灯颜色，输出“停止、等待、通行、未知颜色”。
6. 输入订单状态 `pending/paid/shipped/done`，输出中文状态。
7. 输入文件后缀 `.txt/.jpg/.py/.xlsx`，输出文件类型。
8. 输入运算符 `+ - * /`，输出两个数字的计算结果。
9. 输入 HTTP 状态码，匹配成功、客户端错误、服务器错误和未知状态。
10. 输入一个元组坐标 `(0, 0)`、`(x, 0)`、`(0, y)`、`(x, y)`，判断点的位置。

````python
# match...case：适合一个值对应多个固定分支的情况
# 注意：match...case 需要 Python 3.10 或更高版本

# 1. 命令匹配
command = "start"
match command:
    case "start":
        print("1. 启动程序")
    case "stop":
        print("1. 停止程序")
    case "restart":
        print("1. 重启程序")
    case _:
        print("1. 未知命令")

# 2. 星期数字
weekday = 5
match weekday:
    case 1:
        print("2. 星期一")
    case 2:
        print("2. 星期二")
    case 3:
        print("2. 星期三")
    case 4:
        print("2. 星期四")
    case 5:
        print("2. 星期五")
    case 6:
        print("2. 星期六")
    case 7:
        print("2. 星期日")
    case _:
        print("2. 输入错误")

# 3. 月份所在季度
month = 10
match month:
    case 1 | 2 | 3:
        print("3. 第一季度")
    case 4 | 5 | 6:
        print("3. 第二季度")
    case 7 | 8 | 9:
        print("3. 第三季度")
    case 10 | 11 | 12:
        print("3. 第四季度")
    case _:
        print("3. 月份错误")

# 4. 会员等级折扣
level = "diamond"
match level:
    case "normal":
        discount = 1.0
    case "silver":
        discount = 0.9
    case "gold":
        discount = 0.8
    case "diamond":
        discount = 0.7
    case _:
        discount = 1.0
print("4. 折扣：", discount)

# 5. 交通灯颜色
light = "green"
match light:
    case "red":
        print("5. 停止")
    case "yellow":
        print("5. 等待")
    case "green":
        print("5. 通行")
    case _:
        print("5. 未知颜色")

# 6. 订单状态
status = "paid"
match status:
    case "pending":
        print("6. 待付款")
    case "paid":
        print("6. 已付款")
    case "shipped":
        print("6. 已发货")
    case "done":
        print("6. 已完成")
    case _:
        print("6. 未知状态")

# 7. 文件后缀类型
suffix = ".py"
match suffix:
    case ".txt":
        print("7. 文本文件")
    case ".jpg" | ".png":
        print("7. 图片文件")
    case ".py":
        print("7. Python 文件")
    case ".xlsx":
        print("7. Excel 文件")
    case _:
        print("7. 未知文件类型")

# 8. 简单计算器
a = 20
b = 4
op = "*"
match op:
    case "+":
        result = a + b
    case "-":
        result = a - b
    case "*":
        result = a * b
    case "/":
        # 除法单独处理除数为 0 的情况
        result = a / b if b != 0 else "除数不能为 0"
    case _:
        result = "未知运算符"
print("8. 计算结果：", result)

# 9. HTTP 状态码
code = 500
match code:
    case 200 | 201 | 204:
        print("9. 成功")
    case 400 | 401 | 403 | 404:
        print("9. 客户端错误")
    case 500 | 502 | 503:
        print("9. 服务器错误")
    case _:
        print("9. 未知状态")

# 10. 坐标模式匹配
point = (0, 8)
match point:
    case (0, 0):
        print("10. 原点")
    case (x, 0):
        print(f"10. x 轴上的点，x={x}")
    case (0, y):
        print(f"10. y 轴上的点，y={y}")
    case (x, y):
        # 这里的 x、y 会自动接收元组中的两个值
        print(f"10. 普通点，x={x}, y={y}")
    case _:
        print("10. 不是有效坐标")
````

## 6. 选择哪种条件语句

| 场景 | 推荐写法 |
| --- | --- |
| 只在满足条件时执行一段代码 | `if` |
| 两种互斥情况 | `if...else` |
| 三种及以上区间或条件 | `if...elif...else` |
| 先判断前置条件，再继续判断细节 | 嵌套 `if` |
| 一个值对应多个固定分支 | `match...case` |

写条件语句时最重要的是：先把情况分清楚，再把判断顺序排清楚。

## 7. 常见错误总结

1. 忘记写冒号 `:`。
2. 缩进不一致，尤其是混用了 Tab 和空格。
3. 把判断相等写成赋值，比如把 `==` 写成 `=`。
4. 多分支顺序错误，比如先写 `score >= 60`，后写 `score >= 90`，会导致优秀分数提前进入及格分支。
5. `else` 后面写条件，这是错误的；需要条件时应该用 `elif`。
6. 嵌套太深，导致代码难读，可以考虑合并条件或提前返回。
7. 使用 `match...case` 时忘记 `case _`，导致未知情况没有兜底处理。
