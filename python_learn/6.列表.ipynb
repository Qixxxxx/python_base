# 6. 列表 List

列表是 Python 中最常用的容器类型之一，类型名是 `list`。

核心特点：
- 有序：元素有固定顺序，可以通过下标访问。
- 可变：可以增加、删除、修改元素。
- 可重复：同一个值可以出现多次。
- 可以存放不同类型的数据，但实际开发中通常建议同一个列表里放同一类数据。

## 1. 列表的创建

列表用中括号 `[]` 表示，元素之间用逗号分隔。也可以使用 `list()` 把可迭代对象转换成列表。

```python
nums = [1, 2, 3, 4]
names = ["Tom", "Jerry", "Alice"]
mixed = [1, "hello", True, 3.14]
chars = list("abc")

print(nums)
print(names)
print(mixed)
print(chars)
print(type(nums))
```

### 解释
- 列表用 `[]` 创建，里面的元素按写入顺序保存。
- Python 允许一个列表里放不同类型的数据，但真实项目中通常保持元素类型一致，后续处理更简单。
- `list("abc")` 会把字符串拆成一个个字符，得到 `['a', 'b', 'c']`。

## 2. 下标和切片

列表和字符串一样支持下标、负数下标、切片。切片会生成一个新列表。

```python
letters = ["a", "b", "c", "d", "e"]

print(letters[0])
print(letters[-1])
print(letters[1:4])
print(letters[:3])
print(letters[::2])
print(letters[::-1])
```

### 解释
- 列表下标规则和字符串一致：从 `0` 开始，`-1` 表示最后一个元素。
- 切片得到的是新列表，不是原列表本身。
- `letters[::-1]` 可以快速得到反向列表，但不会修改原列表。

## 3. 修改元素

列表是可变对象，可以通过下标或切片直接修改内容。

```python
scores = [80, 90, 70]
scores[0] = 85
print(scores)

scores[1:3] = [95, 75]
print(scores)

scores[1:2] = [88, 89, 90]
print(scores)
```

### 解释
- 列表是可变对象，所以可以通过下标直接修改某个位置的值。
- 切片赋值可以一次替换多个元素。
- 切片赋值左右两边长度可以不同，因此也能让列表变长或变短。

## 4. 添加元素

| 方法 | 作用 |
| --- | --- |
| `append(x)` | 在末尾添加一个元素 |
| `extend(iterable)` | 在末尾追加多个元素 |
| `insert(index, x)` | 在指定位置插入元素 |

```python
fruits = ["apple", "banana"]

fruits.append("orange")
print(fruits)

fruits.extend(["pear", "grape"])
print(fruits)

fruits.insert(1, "mango")
print(fruits)
```

### 解释
- `append()` 是把参数作为“一个整体”添加到末尾。
- `extend()` 会遍历参数，把里面的元素一个个追加到列表末尾。
- `insert()` 可以指定插入位置，但频繁在列表开头插入会比末尾追加慢。

## 5. 删除元素

| 方法或语句 | 作用 |
| --- | --- |
| `pop()` | 删除并返回最后一个元素 |
| `pop(index)` | 删除并返回指定下标的元素 |
| `remove(x)` | 删除第一个值为 `x` 的元素 |
| `del 列表[index]` | 按下标删除 |
| `clear()` | 清空列表 |

```python
items = ["a", "b", "c", "b", "d"]

last = items.pop()
print(last, items)

second = items.pop(1)
print(second, items)

items.remove("b")
print(items)

del items[0]
print(items)

items.clear()
print(items)
```

### 解释
- `pop()` 会返回被删除的元素，适合“取出并删除”的场景。
- `remove()` 按值删除，而且只删除第一个匹配到的值。
- `clear()` 会保留列表对象本身，但把里面的元素全部清空。

## 6. 查询、统计和成员判断

- `in` / `not in`：判断元素是否存在。
- `index()`：获取元素第一次出现的位置，找不到会报错。
- `count()`：统计元素出现次数。
- `len()`：获取列表长度。

```python
nums = [10, 20, 30, 20, 40]

print(20 in nums)
print(99 not in nums)
print(nums.index(20))
print(nums.count(20))
print(len(nums))
```

### 解释
- `in` 用来判断元素是否存在，写法直观，适合先判断再操作。
- `index()` 找不到元素会报错，所以不确定时可以先用 `in` 判断。
- `count()` 可以统计重复值出现次数，常用于简单计数。

## 7. 排序和反转

- `sort()`：原地排序，会修改原列表。
- `sorted()`：返回一个新列表，不修改原列表。
- `reverse()`：原地反转。
- `reversed()`：返回一个反向迭代器。

```python
nums = [3, 1, 4, 2]

print(sorted(nums))       # 新列表
print(nums)               # 原列表不变

nums.sort()
print(nums)               # 原列表被排序

nums.sort(reverse=True)
print(nums)

nums.reverse()
print(nums)
```

### 解释
- `sorted(nums)` 返回新列表，不影响 `nums` 原来的顺序。
- `nums.sort()` 会原地排序，返回值是 `None`，不要写成 `nums = nums.sort()`。
- `reverse=True` 表示降序排序。

## 8. 按规则排序

`key` 参数可以指定排序规则，常用于按长度、按成绩、按字典中的某个字段排序。

```python
words = ["python", "go", "java", "javascript"]
print(sorted(words, key=len))

students = [("Tom", 80), ("Jerry", 95), ("Alice", 88)]
print(sorted(students, key=lambda item: item[1]))
print(sorted(students, key=lambda item: item[1], reverse=True))
```

### 解释
- `key=len` 表示排序时不直接比较单词本身，而是比较单词长度。
- `lambda item: item[1]` 表示用元组的第二个元素作为排序依据，也就是成绩。
- `key` 非常常用，后面处理字典列表、对象列表时也会频繁见到。

## 9. 遍历列表

直接遍历可以拿到元素；使用 `enumerate()` 可以同时拿到下标和元素。

```python
names = ["Tom", "Jerry", "Alice"]

for name in names:
    print(name)

for index, name in enumerate(names):
    print(index, name)
```

### 解释
- 只需要元素时，直接 `for name in names` 最简洁。
- 同时需要下标和元素时，用 `enumerate()`，比自己维护计数器更稳。
- `enumerate(names, start=1)` 可以让编号从 `1` 开始。

## 10. 列表推导式

列表推导式可以用一行代码生成新列表，适合“遍历 + 处理 + 筛选”的场景。

```python
nums = [1, 2, 3, 4, 5]

squares = [x * x for x in nums]
even_nums = [x for x in nums if x % 2 == 0]
labels = ["偶数" if x % 2 == 0 else "奇数" for x in nums]

print(squares)
print(even_nums)
print(labels)
```

### 解释
- `[x * x for x in nums]` 的含义是：遍历 `nums`，把每个 `x` 平方后放入新列表。
- `if x % 2 == 0` 放在后面表示筛选，只保留满足条件的元素。
- 如果要做“二选一”的结果，条件表达式要写在前面，比如 `A if 条件 else B`。

## 11. 列表复制：引用和浅拷贝

`b = a` 不是复制列表，而是让两个变量指向同一个列表。需要复制时可以使用切片、`copy()` 或 `list()`。

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)       # a 也变了，因为 a 和 b 指向同一个列表

c = a.copy()
c.append(5)
print(a)
print(c)
```

### 解释
- `b = a` 只是多了一个变量名，两个变量指向同一个列表对象。
- `copy()` 会复制出一个新的外层列表，所以给 `c` 添加元素不会影响 `a`。
- 这里讲的是浅拷贝，如果列表里面还有列表，内部列表仍可能被共享。

## 12. 嵌套列表和常见坑

二维列表常用于表示表格或矩阵。创建二维列表时，不要用 `[[0] * 3] * 3`，因为里面的三个小列表其实是同一个对象。

```python
wrong = [[0] * 3] * 3
wrong[0][0] = 1
print(wrong)

right = [[0] * 3 for _ in range(3)]
right[0][0] = 1
print(right)
```

### 解释
- `[[0] * 3] * 3` 看起来创建了三行，实际三行引用的是同一个内部列表。
- 修改 `wrong[0][0]` 时，其他行的第一个元素也跟着变，因为它们本质上是同一个对象。
- 用列表推导式每次重新创建内部列表，才能得到真正独立的三行。
