# 12. 函数

函数用来把一段可以重复使用的代码封装起来。以后需要执行这段逻辑时，不必重新写一遍，只要调用函数即可。

本节重点内容：

- 函数的定义和调用
- 参数：位置参数、关键字参数、默认参数
- 返回值：`return`、多个返回值、没有返回值时的 `None`
- 变量作用域：局部变量、全局变量、`global`
- 函数文档字符串：`docstring`
- 不定长参数：`*args` 和 `**kwargs`
- 函数作为数据传递、`lambda` 匿名函数
- 闭包和递归的基础写法
- 常见错误总结

本章不包含练习题，只保留讲解、示例和详细注释。

## 1. 为什么需要函数

如果一段逻辑会被多次使用，就应该优先考虑封装成函数。函数可以减少重复代码，让程序结构更清晰，也方便后续修改。

```python
# 不使用函数：同一种计算逻辑写了两遍
# 如果以后折扣计算规则变化，需要同时修改多处代码，容易漏改。

prices_1 = [30, 59, 120]
discount_1 = 0.85
total_1 = 0

for price in prices_1:
    # 把每个商品打折后的价格累加到总价中
    total_1 += price * discount_1

print('第一组商品总价：', total_1)

prices_2 = [99, 199]
discount_2 = 0.9
total_2 = 0

for price in prices_2:
    # 这里和上面的逻辑几乎一样，属于重复代码
    total_2 += price * discount_2

print('第二组商品总价：', total_2)


# 使用函数：把通用逻辑封装起来
def calculate_total(prices, discount):
    # prices 表示商品价格列表
    # discount 表示折扣，例如 0.85 表示八五折
    total = 0

    for price in prices:
        # 每次循环处理一个商品价格
        total += price * discount

    # return 把计算结果交还给函数调用者
    return total


print('函数计算第一组：', calculate_total([30, 59, 120], 0.85))
print('函数计算第二组：', calculate_total([99, 199], 0.9))
```

### 解释

- `def calculate_total(prices, discount):` 表示定义一个函数。
- `prices` 和 `discount` 是参数，用来接收调用函数时传进来的数据。
- 函数内部的 `total` 只在函数内部使用，外部不需要关心它的计算过程。
- `return total` 会把最终结果返回，调用者可以直接打印，也可以保存到变量中。

## 2. 函数的定义和调用

函数必须先定义，再调用。只定义函数不会执行函数体，只有写了函数名加小括号，函数才会真正运行。

```python
# 定义函数的基本语法：
# def 函数名():
#     函数体

def say_hello():
    # 这里是函数体，必须缩进
    # 函数体里的代码不会在定义时执行
    print('你好，欢迎学习 Python 函数')


# 调用函数：函数名后面加小括号
say_hello()


def print_line():
    # 这个函数专门负责打印分隔线
    print('-' * 30)


print_line()
print('上面和下面之间有一条分隔线')
print_line()
```

### 解释

- `def` 是定义函数的关键字。
- 函数名建议使用小写字母加下划线，例如 `say_hello`、`print_line`。
- 函数定义行最后必须写冒号 `:`。
- 函数体必须缩进，通常使用 4 个空格。
- `say_hello` 只是函数对象本身，`say_hello()` 才是调用函数。

## 3. 参数：把外部数据传进函数

参数可以让函数变得更灵活。同一个函数，传入不同数据，就能得到不同结果。

```python
def introduce(name, age, city):
    # name、age、city 是形参：定义函数时写在小括号里的变量
    # 它们用于接收函数调用时传入的真实数据
    print(f'大家好，我叫{name}，今年{age}岁，来自{city}')


# 位置参数：按照形参顺序依次传入
introduce('小明', 18, '上海')

# 关键字参数：明确指定参数名，顺序可以调整
introduce(age=20, city='北京', name='小红')


def add(a, b):
    # a 和 b 只是函数内部使用的名字
    # 调用时传入的两个值，会分别交给 a 和 b
    result = a + b
    return result


print(add(10, 20))
print(add(a=3, b=5))
```

### 解释

- 形参：定义函数时写的参数名，例如 `name`、`age`。
- 实参：调用函数时传入的真实数据，例如 `'小明'`、`18`。
- 位置参数依赖顺序，顺序写错就会导致含义错误。
- 关键字参数直接写参数名，代码更清楚，尤其适合参数较多的函数。

## 4. 默认参数

默认参数可以给参数设置默认值。调用函数时如果没有传这个参数，就使用默认值；如果传了，就使用传入的新值。

```python
def make_order(product, count=1, price=10):
    # product 没有默认值，调用时必须传入
    # count 默认是 1，表示默认购买 1 件
    # price 默认是 10，表示默认单价 10 元
    total = count * price
    print(f'商品：{product}，数量：{count}，单价：{price}，总价：{total}')


make_order('矿泉水')
make_order('笔记本', count=3)
make_order('键盘', count=2, price=99)


# 注意：默认参数尽量不要直接使用可变对象，例如列表、字典。
# 推荐使用 None 作为默认值，再在函数内部创建新的列表。
def add_student(name, students=None):
    # 第一次调用或没有传入列表时，students 是 None
    if students is None:
        # 每次需要新列表时，都在函数内部重新创建
        students = []

    students.append(name)
    return students


print(add_student('Tom'))
print(add_student('Jerry'))

class_1 = []
print(add_student('Alice', class_1))
print(add_student('Bob', class_1))
```

### 解释

- 默认参数必须写在普通参数后面，例如 `def f(a, b=1)`。
- 调用 `make_order('矿泉水')` 时，`count` 和 `price` 使用默认值。
- 默认参数可以减少重复传参，让函数调用更简洁。
- 默认值如果是列表、字典这类可变对象，容易造成多次调用共享同一个对象，因此更推荐使用 `None`。

## 5. 返回值：把结果交还给调用者

`return` 用来结束函数，并把结果返回给调用函数的位置。函数可以返回一个值，也可以返回多个值。

```python
def add(a, b):
    # 计算两个数的和
    result = a + b
    # 把计算结果返回出去
    return result


sum_result = add(10, 20)
print('加法结果：', sum_result)


def check_score(score):
    # return 会立即结束函数
    # 如果 score >= 60 成立，下面这个 return 执行后，函数就结束了
    if score >= 60:
        return '及格'

    # 只有上面的条件不成立时，才会执行到这里
    return '不及格'


print(check_score(85))
print(check_score(45))


def get_user_info():
    # Python 可以一次返回多个值
    # 本质上返回的是一个元组，只是可以直接拆包接收
    return 'Tom', 18, '上海'


name, age, city = get_user_info()
print(name, age, city)


def only_print():
    # 这个函数只有打印，没有显式 return
    print('我只是负责打印')


result = only_print()
# 没有显式 return 的函数，默认返回 None
print('only_print 的返回值：', result)
```

### 解释

- `print()` 只是把内容显示到屏幕上，不等于返回值。
- `return` 才是把结果交给函数外部继续使用。
- 一个函数执行到 `return` 后会立刻结束，后面的代码不会继续执行。
- 没有写 `return` 或只写 `return` 不带值时，函数返回 `None`。

## 6. 变量作用域

作用域指变量能够被访问的范围。函数内部定义的变量叫局部变量，通常只能在函数内部使用；函数外部定义的变量叫全局变量。

```python
message = '我是全局变量'


def show_scope():
    # local_message 是局部变量，只能在函数内部直接使用
    local_message = '我是局部变量'

    # 函数内部可以读取全局变量
    print(message)
    print(local_message)


show_scope()

# print(local_message)
# 上面这行如果取消注释，会报 NameError，因为 local_message 只存在于函数内部。


counter = 0


def add_one():
    # 如果要在函数内部修改全局变量，需要使用 global 声明
    # 真实项目中不建议频繁修改全局变量，因为会让数据变化难追踪
    global counter
    counter += 1


add_one()
add_one()
print('counter =', counter)
```

### 解释

- 局部变量在函数调用时创建，函数结束后通常就不能再访问。
- 函数内部可以读取全局变量，但直接给全局变量赋值时，会被当作创建局部变量。
- `global` 可以声明要修改的是全局变量。
- 初学阶段要记住：优先使用参数和返回值传递数据，少依赖全局变量。

## 7. 文档字符串 docstring

文档字符串写在函数体第一行，用三引号包裹，用来说明函数的用途、参数和返回值。

```python
def calculate_area(width, height):
    '''
    计算矩形面积。

    参数：
    width：矩形宽度
    height：矩形高度

    返回值：
    宽度乘以高度得到的面积
    '''
    return width * height


area = calculate_area(5, 3)
print('面积：', area)

# __doc__ 可以查看函数的文档字符串
print(calculate_area.__doc__)

# help(calculate_area)
# 在普通 Python 环境中，可以使用 help() 查看函数说明。
```

### 解释

- 普通注释主要给阅读代码的人看。
- `docstring` 可以被工具读取，例如 `help()`、编辑器提示、自动文档工具。
- 重要函数建议写清楚：函数做什么、参数是什么、返回什么。

## 8. 不定长参数：`*args` 和 `**kwargs`

当函数需要接收不确定数量的参数时，可以使用不定长参数。`*args` 接收多个位置参数，`**kwargs` 接收多个关键字参数。

```python
def sum_numbers(*args):
    # *args 会把多个位置参数打包成一个元组
    # 例如 sum_numbers(1, 2, 3) 中，args 就是 (1, 2, 3)
    total = 0

    for num in args:
        total += num

    return total


print(sum_numbers(1, 2, 3))
print(sum_numbers(10, 20, 30, 40))


def print_profile(**kwargs):
    # **kwargs 会把多个关键字参数打包成一个字典
    # 例如 name='Tom', age=18 会变成 {'name': 'Tom', 'age': 18}
    for key, value in kwargs.items():
        print(key, value)


print_profile(name='Tom', age=18, city='上海')


def register_user(name, age, *hobbies, **extra):
    # 普通参数 name、age 必须传
    # *hobbies 接收额外的位置参数
    # **extra 接收额外的关键字参数
    print('姓名：', name)
    print('年龄：', age)
    print('爱好：', hobbies)
    print('其他信息：', extra)


register_user('Alice', 20, '阅读', '跑步', city='北京', vip=True)
```

### 解释

- `args` 和 `kwargs` 只是习惯命名，关键是前面的 `*` 和 `**`。
- `*args` 得到的是元组，适合处理一组数量不固定的位置数据。
- `**kwargs` 得到的是字典，适合处理一组数量不固定的命名数据。
- 参数顺序通常是：普通参数、默认参数、`*args`、`**kwargs`。

## 9. 函数也可以作为数据传递

在 Python 中，函数本身也是对象，可以保存到变量中，也可以作为参数传给另一个函数。

```python
def square(num):
    # 返回平方
    return num * num


def cube(num):
    # 返回立方
    return num * num * num


def apply_func(nums, func):
    # nums 是数字列表
    # func 是一个函数，用来决定每个数字如何被处理
    result = []

    for num in nums:
        # func(num) 表示调用传进来的函数
        result.append(func(num))

    return result


numbers = [1, 2, 3, 4]

print(apply_func(numbers, square))
print(apply_func(numbers, cube))


students = [
    {'name': 'Tom', 'score': 80},
    {'name': 'Jerry', 'score': 95},
    {'name': 'Alice', 'score': 88}
]

# sorted() 的 key 参数需要接收一个函数
# 这个函数告诉 sorted()：排序时应该看每个学生字典里的哪个值
sorted_students = sorted(students, key=lambda item: item['score'], reverse=True)
print(sorted_students)
```

### 解释

- `apply_func(numbers, square)` 传入的是函数 `square` 本身，不要写成 `square()`。
- `square()` 表示立刻调用函数；`square` 表示把函数对象传过去。
- `sorted(..., key=...)` 是函数作为参数的常见应用。

## 10. `lambda` 匿名函数

`lambda` 用来定义非常短的小函数，通常只写一行，适合临时传给 `sorted()`、`map()`、`filter()` 等函数使用。

```python
# 普通函数写法
def double(num):
    return num * 2


# lambda 写法：参数在冒号左边，返回表达式在冒号右边
double_lambda = lambda num: num * 2

print(double(5))
print(double_lambda(5))


numbers = [1, 2, 3, 4, 5]

# map() 会把 lambda 应用到列表中的每个元素上
squares = list(map(lambda x: x * x, numbers))
print('平方列表：', squares)

# filter() 会保留 lambda 返回 True 的元素
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print('偶数列表：', even_numbers)


words = ['python', 'go', 'javascript', 'java']

# 按字符串长度排序
sorted_words = sorted(words, key=lambda word: len(word))
print(sorted_words)
```

### 解释

- `lambda num: num * 2` 等价于一个只返回 `num * 2` 的简单函数。
- `lambda` 适合短逻辑，不适合写复杂流程。
- 如果逻辑需要多行、需要注释、需要复用，优先使用 `def` 定义普通函数。

## 11. 闭包基础

如果内部函数使用了外部函数的变量，并且外部函数把内部函数返回出去，就形成了闭包。闭包可以记住外层函数当时的状态。

```python
def make_multiplier(factor):
    # factor 是外部函数的局部变量
    # 它会被内部函数 multiplier 使用

    def multiplier(num):
        # 内部函数可以访问外部函数中的变量 factor
        return num * factor

    # 返回的是函数本身，不是调用结果
    return multiplier


double = make_multiplier(2)
triple = make_multiplier(3)

# double 记住了 factor = 2
print(double(10))

# triple 记住了 factor = 3
print(triple(10))
```

### 解释

- `make_multiplier(2)` 返回了内部函数 `multiplier`。
- 虽然 `make_multiplier` 已经执行结束，但返回出来的内部函数仍然记得 `factor` 的值。
- 闭包是装饰器的基础，后面学习装饰器时会再次用到。

## 12. 递归函数

函数内部调用自己，叫递归。递归一定要有结束条件，否则会无限调用，直到程序报错。

```python
def factorial(n):
    # 递归示例：计算 n 的阶乘
    # 5! = 5 * 4 * 3 * 2 * 1

    if n < 0:
        # 负数没有普通意义上的阶乘，这里返回 None 表示无法计算
        return None

    if n == 0 or n == 1:
        # 这是递归结束条件
        # 如果没有这个条件，函数会一直调用自己
        return 1

    # 把大问题拆成小问题：n! = n * (n - 1)!
    return n * factorial(n - 1)


print(factorial(5))
print(factorial(1))
print(factorial(-3))
```

### 解释

- 递归包含两个关键部分：结束条件、递归调用。
- `factorial(5)` 会变成 `5 * factorial(4)`，再继续向下拆分。
- 当递归调用到 `factorial(1)` 时触发结束条件，开始一层层返回结果。
- 初学时不要滥用递归，很多问题用循环更直观。

## 13. 函数设计建议

| 建议 | 说明 |
| --- | --- |
| 函数名表达动作 | 例如 `calculate_total`、`get_user_info`、`check_score` |
| 一个函数做一件事 | 不要把输入、计算、打印、保存全部塞进一个函数 |
| 优先使用参数和返回值 | 少依赖全局变量，数据流更清楚 |
| 复杂逻辑写注释 | 注释解释为什么这样做，而不是重复代码表面意思 |
| 重要函数写 docstring | 方便自己和别人快速理解函数用途 |
| 不要让函数太长 | 函数过长时，可以继续拆成更小的函数 |

函数写得好，代码会像积木一样可以组合；函数写得太杂，后期修改会很痛苦。

## 14. 常见错误总结

1. 只定义函数，忘记调用函数。
2. 调用函数时参数数量不匹配。
3. 位置参数顺序写错，导致数据含义错乱。
4. 把 `print()` 当成 `return` 使用，导致外部拿不到结果。
5. `return` 后面继续写希望执行的代码，但实际上不会执行。
6. 在函数外部访问函数内部的局部变量。
7. 在函数内部修改全局变量却忘记写 `global`。
8. 默认参数使用列表或字典，导致多次调用共享同一个对象。
9. 把函数对象和函数调用混淆，例如该传 `square` 时写成 `square()`。
10. 递归函数没有结束条件，导致无限递归。

```python
# 常见错误示例说明：这些代码以注释形式保留，避免直接运行时报错。

# 1. 忘记调用函数
def hello():
    print('hello')

# hello      # 只是函数对象，不会执行
hello()      # 加上小括号才会调用


# 2. 参数数量不匹配
def add(a, b):
    return a + b

# add(1)     # 错误：少传了一个参数 b
# add(1, 2, 3)  # 错误：多传了一个参数


# 3. print 不等于 return
def wrong_add(a, b):
    # 这个函数只是打印结果，没有返回结果
    print(a + b)


value = wrong_add(1, 2)
print('wrong_add 的返回值：', value)


def right_add(a, b):
    # 这个函数把结果返回出去，外部可以继续使用
    return a + b


value = right_add(1, 2)
print('right_add 的返回值：', value)
```
