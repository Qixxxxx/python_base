# 8. 字典 Dictionary

字典是用来保存“键值对”的容器，类型名是 `dict`。

核心特点：
- 通过键 `key` 查找值 `value`。
- 可变：可以增加、删除、修改键值对。
- 键必须唯一；同一个键重复出现时，后面的值会覆盖前面的值。
- 键必须是不可变且可哈希的对象，比如字符串、数字、元组。

## 1. 字典的创建

字典使用花括号 `{}` 表示，每一项都是 `key: value`。

```python
student = {"name": "小明", "age": 18, "city": "北京"}
empty = {}

user = dict(name="Tom", age=20)
pairs = dict([("a", 1), ("b", 2)])

print(student)
print(empty)
print(user)
print(pairs)
print(type(student))
```

### 解释
- 字典保存的是键值对，适合表达“某个名字对应某个值”的数据。
- `dict(name="Tom", age=20)` 这种写法要求键名能写成合法标识符。
- `dict([("a", 1), ("b", 2)])` 常用于把二元组列表转换成字典。

## 2. 访问字典中的值

可以使用 `字典[key]` 或 `get()` 获取值。

- `字典[key]`：键不存在会报错。
- `get(key, 默认值)`：键不存在时返回默认值。

```python
student = {"name": "小明", "age": 18}

print(student["name"])
print(student.get("age"))
print(student.get("score"))
print(student.get("score", 0))

# print(student["score"])   # KeyError：键不存在
```

### 解释
- 当确定键一定存在时，用 `student["name"]` 很直接。
- 当不确定键是否存在时，用 `get()` 更安全，因为不会抛出 `KeyError`。
- `get("score", 0)` 的 `0` 是默认值，表示没有成绩时按 0 处理。

## 3. 增加和修改键值对

给一个不存在的键赋值，就是新增；给一个已存在的键赋值，就是修改。

```python
student = {"name": "小明", "age": 18}

student["score"] = 95       # 新增
student["age"] = 19         # 修改
student.update({"city": "上海", "score": 98})

print(student)
```

### 解释
- 字典赋值语法同时承担“新增”和“修改”两个功能。
- 如果键不存在，`student["score"] = 95` 会新增键值对。
- 如果键已存在，重新赋值会覆盖旧值；`update()` 适合一次合并多项数据。

## 4. 删除键值对

| 方法或语句 | 作用 |
| --- | --- |
| `pop(key)` | 删除指定键，并返回对应值 |
| `pop(key, 默认值)` | 键不存在时返回默认值，避免报错 |
| `popitem()` | 删除并返回最后插入的键值对 |
| `del 字典[key]` | 删除指定键 |
| `clear()` | 清空字典 |

```python
info = {"name": "Tom", "age": 20, "city": "北京"}

age = info.pop("age")
print(age, info)

score = info.pop("score", 0)
print(score, info)

del info["city"]
print(info)

info.clear()
print(info)
```

### 解释
- `pop("age")` 删除键的同时返回被删除的值。
- `pop("score", 0)` 给了默认值，所以键不存在时不会报错。
- `del info["city"]` 也能删除，但键不存在会报错，适合确定键存在的情况。

## 5. 判断键是否存在

`in` 判断的是字典的键，不是值。

```python
student = {"name": "小明", "age": 18}

print("name" in student)
print("小明" in student)
print("score" not in student)
```

### 解释
- 对字典使用 `in` 时，检查的是键，不是值。
- 所以 `"name" in student` 是 `True`，但 `"小明" in student` 是 `False`。
- 如果要判断值是否存在，需要写 `"小明" in student.values()`。

## 6. 遍历字典

常见遍历方式：遍历键、遍历值、同时遍历键和值。

```python
student = {"name": "小明", "age": 18, "score": 95}

for key in student:
    print(key, student[key])

for value in student.values():
    print(value)

for key, value in student.items():
    print(key, value)
```

### 解释
- 直接遍历字典时，默认拿到的是键。
- `values()` 只遍历值，不关心键。
- `items()` 会同时拿到键和值，是遍历字典时最常用的方式之一。

## 7. `keys()`、`values()`、`items()`

这三个方法返回的是视图对象，不是普通列表。如果需要列表，可以再套一层 `list()`。

```python
student = {"name": "小明", "age": 18, "score": 95}

print(student.keys())
print(student.values())
print(student.items())

print(list(student.keys()))
print(list(student.items()))
```

### 解释
- `keys()`、`values()`、`items()` 返回的是视图对象，可以遍历，但不是普通列表。
- 视图对象会跟随字典变化而变化。
- 如果需要下标访问、切片或固定快照，可以用 `list()` 转成列表。

## 8. `setdefault()`

`setdefault(key, 默认值)`：如果键存在，返回原来的值；如果键不存在，新增这个键并设置默认值。

```python
scores = {"Tom": [80, 90]}

scores.setdefault("Tom", []).append(95)
scores.setdefault("Jerry", []).append(88)

print(scores)
```

### 解释
- `setdefault("Tom", [])`：如果 `Tom` 已存在，就返回原来的列表。
- `setdefault("Jerry", [])`：如果 `Jerry` 不存在，就先创建一个空列表再返回它。
- 这种写法常用于“按类别分组收集数据”。

## 9. 字典推导式

字典推导式可以快速生成字典，语法类似列表推导式。

```python
nums = [1, 2, 3, 4]
squares = {x: x * x for x in nums}
even_squares = {x: x * x for x in nums if x % 2 == 0}

print(squares)
print(even_squares)

names = ["Tom", "Jerry", "Alice"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)
```

### 解释
- `{x: x * x for x in nums}` 会把每个数字作为键，把平方结果作为值。
- 字典推导式的左侧必须写成 `key: value` 的形式。
- 如果多个元素生成了相同的键，后生成的值会覆盖前面的值。

## 10. 字典合并

`update()` 会修改原字典；`|` 会生成一个新字典。相同键以后面的值为准。

```python
a = {"name": "Tom", "age": 18}
b = {"age": 20, "city": "北京"}

c = a | b
print(c)
print(a)

a.update(b)
print(a)
```

### 解释
- `a | b` 会生成新字典，`a` 和 `b` 原本的内容不变。
- `a.update(b)` 会直接修改 `a`。
- 合并时如果有相同键，右边字典的值会覆盖左边字典的值。

## 11. 嵌套字典

字典的值可以继续是字典，常用于表示结构化数据。

```python
students = {
    "1001": {"name": "小明", "score": 90},
    "1002": {"name": "小红", "score": 95}
}

print(students["1001"]["name"])
print(students.get("1003", {}).get("name", "查无此人"))
```

### 解释
- 嵌套字典适合表示学生、商品、配置等结构化信息。
- `students["1001"]["name"]` 是先根据学号找到学生字典，再从学生字典里取姓名。
- `get("1003", {})` 给了一个空字典默认值，可以避免第一层键不存在时报错。

## 12. 按键或按值排序

字典本身保存键值关系。排序时通常把 `items()` 转成列表，再用 `sorted()`。

```python
scores = {"Tom": 80, "Jerry": 95, "Alice": 88}

print(sorted(scores.items()))
print(sorted(scores.items(), key=lambda item: item[1]))
print(sorted(scores.items(), key=lambda item: item[1], reverse=True))
```

### 解释
- `scores.items()` 得到类似 `[('Tom', 80), ...]` 的键值对结构。
- `sorted()` 排序后的结果是列表，不再是字典。
- `key=lambda item: item[1]` 表示按分数排序，而不是按姓名排序。
