# 7. 元组 Tuple

元组和列表很像，都是有序容器。区别是：列表可变，元组不可变。

核心特点：
- 有序：支持下标和切片。
- 不可变：不能增加、删除、修改元素。
- 可重复：同一个值可以出现多次。
- 通常用于保存不希望被修改的一组数据。

## 1. 元组的创建

元组通常用小括号 `()` 表示。创建只有一个元素的元组时，必须在元素后面加逗号。

```python
t1 = (1, 2, 3)
t2 = "Tom", 18, "北京"      # 小括号可以省略，叫元组打包
empty = ()
single = (1,)
not_tuple = (1)

print(t1)
print(t2)
print(type(single))
print(type(not_tuple))
```

### 解释
- 元组真正的关键不是小括号，而是逗号；`(1,)` 才是单元素元组。
- `t2 = "Tom", 18, "北京"` 这种写法叫元组打包，Python 会自动把多个值打包成元组。
- `(1)` 只是普通整数加了一层括号，所以类型仍然是 `int`。

## 2. 下标和切片

元组是序列类型，所以也支持下标、负数下标、切片。

```python
colors = ("red", "green", "blue", "black")

print(colors[0])
print(colors[-1])
print(colors[1:3])
print(colors[::-1])
```

### 解释
- 元组属于序列类型，所以访问方式和列表、字符串很像。
- 切片得到的新结果仍然是元组。
- 元组不能修改，但可以通过切片读取其中一部分。

## 3. 元组不可变

元组不能修改元素，不能追加元素，也不能删除其中某个元素。

```python
point = (10, 20)
# point[0] = 99       # TypeError：元组不支持修改

new_point = (99,) + point[1:]
print(point)
print(new_point)
```

### 解释
- 元组不可变，所以不能直接替换某个下标位置的元素。
- 如果确实想得到变化后的结果，只能创建一个新元组。
- 这种特性适合保存坐标、日期、配置项等不希望被随意修改的数据。

## 4. 遍历、长度和成员判断

元组可以使用 `for` 遍历，也可以使用 `len()`、`in`、`not in`。

```python
nums = (10, 20, 30)

print(len(nums))
print(20 in nums)

for num in nums:
    print(num)
```

### 解释
- 元组可以遍历，说明它是可迭代对象。
- `in` 判断某个值是否存在于元组中。
- 元组经常用于函数返回值，遍历时可以像列表一样处理。

## 5. 元组的两个常用方法

元组的方法很少，因为它不可变。

| 方法 | 作用 |
| --- | --- |
| `count(x)` | 统计元素出现次数 |
| `index(x)` | 查找元素第一次出现的位置 |

```python
data = (1, 2, 3, 2, 4, 2)

print(data.count(2))
print(data.index(3))
```

### 解释
- 因为元组不可变，所以没有 `append()`、`remove()`、`sort()` 这类会修改自身的方法。
- `count()` 只负责统计，不修改元组。
- `index()` 只负责查找，不修改元组；找不到时会报错。

## 6. 元组解包

元组解包可以把元组中的元素一次性赋值给多个变量。变量个数通常要和元素个数一致。

```python
student = ("小明", 18, "北京")
name, age, city = student

print(name)
print(age)
print(city)

a = 10
b = 20
a, b = b, a
print(a, b)
```

### 解释
- 解包时，左边变量数量通常要和右边元素数量一致。
- `name, age, city = student` 比用下标逐个取值更清晰。
- `a, b = b, a` 是 Python 交换两个变量的常见写法，不需要临时变量。

## 7. 星号解包

如果只关心一部分元素，可以用 `*变量名` 接收剩下的元素。星号变量接收到的是列表。

```python
nums = (1, 2, 3, 4, 5)

first, *middle, last = nums
print(first)
print(middle)
print(last)
```

### 解释
- 星号变量会接收“剩下的所有元素”，接收到的结果是列表。
- 一个解包表达式里通常只能有一个星号变量，否则 Python 不知道如何分配元素。
- 这种写法适合只关心开头、结尾，中间数量不固定的情况。

## 8. 不可变不等于里面的对象都不能变

元组本身不能改：不能让某个位置指向新对象。但如果元组里放的是列表，列表本身仍然可以修改。

```python
t = ([1, 2], "ok")
t[0].append(3)

print(t)
# t[0] = [9, 9]       # TypeError：不能替换元组中的元素
```

### 解释
- 元组不可变指的是：元组中每个位置不能改为指向别的对象。
- 如果某个位置本来指向的是列表，这个列表对象自身仍然可以变化。
- 所以“元组不可变”不等于“里面所有层级的数据都绝对不变”。

## 9. 元组和列表的选择

| 类型 | 是否可变 | 适合场景 |
| --- | --- | --- |
| `list` | 可变 | 数据需要频繁增删改 |
| `tuple` | 不可变 | 数据结构固定，不希望被修改 |

简单理解：会变就用列表，不该变就用元组。

## 10. 元组的常见用途

- 函数返回多个值。
- 保存固定结构的数据，比如坐标 `(x, y)`。
- 作为字典的键，前提是元组里的元素也都是不可变对象。
- 配合解包让代码更简洁。

```python
def get_user():
    return "Tom", 18

name, age = get_user()
print(name, age)

locations = {
    (0, 0): "原点",
    (1, 2): "目标点"
}
print(locations[(1, 2)])
```

### 解释
- Python 函数返回多个值时，本质上经常是返回一个元组。
- `(0, 0)`、`(1, 2)` 这种坐标结构固定，很适合用元组表示。
- 元组能作为字典键的前提是：元组内部的元素也都必须可哈希。
