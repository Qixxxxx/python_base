## 1. `dtype` 数据类型

NumPy 数组中的元素通常是同一种类型。常见类型包括整数、浮点数、布尔值、字符串。


```python
import numpy as np

int_arr = np.array([1, 2, 3])
float_arr = np.array([1.0, 2.5, 3.8])
bool_arr = np.array([True, False, True])

print(int_arr.dtype)
print(float_arr.dtype)
print(bool_arr.dtype)

# 指定 dtype 可以控制数据类型。
small_int = np.array([1, 2, 3], dtype=np.int16)
print(small_int.dtype)
print(small_int.itemsize)

# 混合整数和浮点数时，会自动转换成浮点数。
mixed = np.array([1, 2.5, 3])
print(mixed)
print(mixed.dtype)
```

### 解释

- `int64`、`float64` 是常见默认类型。
- 类型占用内存不同，例如 `int16` 比 `int64` 更省内存，但可表示范围更小。
- 数组类型统一后，底层计算更高效。


## 2. 类型转换：`astype`

`astype()` 会返回转换类型后的新数组，不会直接修改原数组。


```python
import numpy as np

arr = np.array([1.2, 2.8, 3.5])

# 转换成整数会直接截断小数部分，不是四舍五入。
int_arr = arr.astype(int)
print(int_arr)
print(arr)

# 转成字符串。
str_arr = arr.astype(str)
print(str_arr)
print(str_arr.dtype)

# 布尔数组转整数，True 变 1，False 变 0。
bool_arr = np.array([True, False, True])
print(bool_arr.astype(int))
```

### 解释

- `astype` 常用于读取 CSV 后转换数据类型。
- 浮点转整数会截断小数，必要时先使用 `round`。
- 转换失败会报错，例如无法把 `'abc'` 转成整数。


## 3. 视图和拷贝

视图和原数组共享数据，拷贝拥有独立数据。理解这一点很重要，否则容易误改原数组。


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

# 切片通常返回视图。
view = arr[1:4]
view[0] = 999
print('修改视图后的原数组：', arr)

# copy 返回独立拷贝。
arr2 = np.array([1, 2, 3, 4, 5])
copy_arr = arr2[1:4].copy()
copy_arr[0] = 888
print('修改拷贝后的原数组：', arr2)
print('拷贝数组：', copy_arr)

# base 可以帮助判断数组是否可能共享底层数据。
print(view.base is arr)
print(copy_arr.base is arr2)
```

### 解释

- 切片视图节省内存，但修改时会影响原数组。
- 如果希望安全地独立修改，使用 `.copy()`。
- `ravel()` 也可能返回视图，`flatten()` 返回拷贝。


## 4. 缺失值和特殊值

NumPy 中常见特殊值：

| 值 | 含义 | 判断方法 |
| --- | --- | --- |
| `np.nan` | Not a Number，常表示缺失值 | `np.isnan()` |
| `np.inf` | 正无穷 | `np.isinf()` |
| `-np.inf` | 负无穷 | `np.isinf()` |
| 有限数 | 不是 NaN，也不是无穷 | `np.isfinite()` |


```python
import numpy as np

arr = np.array([1.0, np.nan, 2.5, np.inf, -np.inf, 4.0])

print('原数组：', arr)

# 判断哪些位置是 NaN。
print('isnan：', np.isnan(arr))

# 判断哪些位置是无穷。
print('isinf：', np.isinf(arr))

# 判断哪些位置是有限数字。
print('isfinite：', np.isfinite(arr))

# 过滤掉 NaN 和无穷，只保留有限数字。
finite_arr = arr[np.isfinite(arr)]
print('有限数字：', finite_arr)

# nanmean 会忽略 NaN，但不会忽略 inf。
arr_with_nan = np.array([1.0, np.nan, 3.0])
print('普通 mean：', np.mean(arr_with_nan))
print('nanmean：', np.nanmean(arr_with_nan))

# nan_to_num 可以把 NaN 和无穷替换成指定值。
clean = np.nan_to_num(arr, nan=0.0, posinf=999.0, neginf=-999.0)
print('替换后：', clean)
```

### 解释

- `np.nan == np.nan` 的结果是 False，所以不能用 `==` 判断 NaN。
- 判断 NaN 应使用 `np.isnan()`。
- 统计时如果有 NaN，普通 `mean`、`sum` 可能得到 NaN。
- `np.nanmean`、`np.nansum` 等函数可以忽略 NaN。


## 5. 本章练习题

1. 创建浮点数组 `[1.1, 2.2, 3.3]`，转换成整数数组。
2. 创建数组 `[1, 2, 3, 4]`，切片取中间两个元素，修改切片并观察原数组。
3. 使用 `.copy()` 重做第 2 题，让原数组不受影响。
4. 创建含有 `np.nan` 的数组，统计非缺失值平均数。
5. 把数组中的 NaN 替换成 0。


```python
import numpy as np

# 1. 类型转换
arr = np.array([1.1, 2.2, 3.3])
print(arr.astype(int))

# 2. 切片视图会影响原数组
arr2 = np.array([1, 2, 3, 4])
view = arr2[1:3]
view[0] = 99
print(arr2)

# 3. copy 不影响原数组
arr3 = np.array([1, 2, 3, 4])
copy_part = arr3[1:3].copy()
copy_part[0] = 88
print(arr3)
print(copy_part)

# 4. 忽略 NaN 计算平均数
scores = np.array([80, np.nan, 90, 70])
print(np.nanmean(scores))

# 5. 替换 NaN
print(np.nan_to_num(scores, nan=0))
```

## 6. 常见错误总结

1. 误以为切片一定是新数组。
2. 浮点转整数时忽略小数会被截断。
3. 用 `== np.nan` 判断缺失值。
4. 含 NaN 的数组直接做普通平均数。
5. 为了省内存使用小 dtype，却导致数值范围不够。
