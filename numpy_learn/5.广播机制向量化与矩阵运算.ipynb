## 1. 广播机制

广播允许不同形状的数组在满足规则时一起运算。它可以避免手写循环，让代码更简洁。


```python
import numpy as np

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 标量会广播到每个元素。
print(matrix + 10)

# 一维数组 shape=(3,) 可以广播到每一行。
col_values = np.array([10, 20, 30])
print(matrix + col_values)

# 如果想按列给每一行加不同值，需要把数组变成 2 行 1 列。
row_values = np.array([100, 200]).reshape(2, 1)
print(matrix + row_values)
```

### 解释

- 广播会从最后一个维度开始比较形状。
- 两个维度相等，或其中一个为 1，就可以广播。
- `reshape(2, 1)` 常用于把一维数组变成列向量。


## 2. 广播常见场景表

| 场景 | 写法 | 说明 |
| --- | --- | --- |
| 所有元素加同一个数 | `arr + 10` | 标量广播 |
| 每列减去一个均值 | `data - data.mean(axis=0)` | 一维数组广播到每一行 |
| 每行减去一个均值 | `data - data.mean(axis=1).reshape(-1, 1)` | 列向量广播到每一列 |
| 标准化 | `(data - mean) / std` | 数据预处理常用 |
| 生成网格 | `x[:, None] + y[None, :]` | 通过增加维度组合数据 |


```python
import numpy as np

data = np.array([
    [86, 78, 92],
    [91, 88, 95],
    [79, 90, 84],
    [88, 76, 81]
], dtype=float)

# 每一列表示一门课。
# 先计算每门课平均分，再让每个学生对应科目减去该科平均分。
subject_mean = data.mean(axis=0)
centered_by_subject = data - subject_mean
print('按科目中心化：')
print(centered_by_subject)

# 每一行表示一个学生。
# mean(axis=1) 得到每个学生平均分，shape=(4,)。
# reshape(-1, 1) 转成 4 行 1 列，才能按行广播。
student_mean = data.mean(axis=1).reshape(-1, 1)
centered_by_student = data - student_mean
print('按学生中心化：')
print(centered_by_student)

# 标准化：每门课减去平均值，再除以标准差。
subject_std = data.std(axis=0)
standardized = (data - subject_mean) / subject_std
print('标准化：')
print(standardized)
```

### 解释

- `data.mean(axis=0)` 得到每列平均值，形状是 `(3,)`。
- `(4, 3)` 和 `(3,)` 可以广播，因为最后一维都是 3。
- `data.mean(axis=1)` 得到形状 `(4,)`，不能直接按行减，需要变成 `(4, 1)`。


## 3. 向量化替代循环

向量化就是用数组整体运算替代 Python 循环。NumPy 的很多速度优势都来自这里。


```python
import numpy as np

prices = np.array([100, 200, 150, 80])
discounts = np.array([0.9, 0.8, 0.85, 0.95])

# 不需要写 for 循环，直接逐元素相乘。
final_prices = prices * discounts
print(final_prices)

# 条件逻辑也可以向量化。
# 价格大于等于 150 的商品再减 20，否则不变。
new_prices = np.where(final_prices >= 150, final_prices - 20, final_prices)
print(new_prices)

# 求总价。
print(np.sum(new_prices))
```

### 解释

- `prices * discounts` 是逐元素乘法。
- `np.where(condition, x, y)` 类似向量化的三元表达式。
- 条件成立取 `x`，否则取 `y`。


## 4. 矩阵运算方法表

| 写法 | 作用 |
| --- | --- |
| `A @ B` | 矩阵乘法 |
| `np.matmul(A, B)` | 矩阵乘法 |
| `np.dot(a, b)` | 点积或矩阵乘法，取决于维度 |
| `A.T` | 转置 |
| `np.linalg.inv(A)` | 逆矩阵 |
| `np.linalg.det(A)` | 行列式 |
| `np.linalg.solve(A, b)` | 解线性方程组 |


```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [10, 20],
    [30, 40]
])

# 注意：A * B 是逐元素乘法，不是矩阵乘法。
print('逐元素乘法：')
print(A * B)

# @ 才是矩阵乘法。
print('矩阵乘法：')
print(A @ B)

# 转置。
print('转置：')
print(A.T)

# 行列式。
print('行列式：', np.linalg.det(A))

# 逆矩阵。
print('逆矩阵：')
print(np.linalg.inv(A))

# 解方程组：
# 2x + y = 8
# x + 3y = 13
coef = np.array([[2, 1], [1, 3]])
value = np.array([8, 13])
solution = np.linalg.solve(coef, value)
print('方程组解：', solution)
```

### 解释

- `*` 是逐元素乘法。
- `@` 是矩阵乘法。
- 矩阵乘法要求左矩阵列数等于右矩阵行数。
- `np.linalg.solve` 比先求逆再相乘更推荐。


## 5. 本章练习题

1. 创建一个 3 行 4 列数组，让每一列分别加 `[10, 20, 30, 40]`。
2. 计算每一行的平均值，并让每一行减去自己的平均值。
3. 使用 `np.where` 把小于 60 的成绩替换成 0。
4. 计算两个 2x2 矩阵的逐元素乘法和矩阵乘法。
5. 解方程组：`x + y = 10`，`2x - y = 5`。


```python
import numpy as np

# 1. 每列加不同值
arr = np.arange(1, 13).reshape(3, 4)
print(arr + np.array([10, 20, 30, 40]))

# 2. 每行减去自己的平均值
row_mean = arr.mean(axis=1).reshape(-1, 1)
print(arr - row_mean)

# 3. 不及格替换成 0
scores = np.array([55, 80, 92, 40, 67])
print(np.where(scores < 60, 0, scores))

# 4. 逐元素乘法和矩阵乘法
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(A * B)
print(A @ B)

# 5. 解方程组
coef = np.array([[1, 1], [2, -1]])
value = np.array([10, 5])
print(np.linalg.solve(coef, value))
```

## 6. 常见错误总结

1. 把 `*` 当作矩阵乘法。
2. 广播形状不匹配，尤其是 `(n,)` 和 `(n, 1)` 混淆。
3. 标准化时忘记指定 `axis`。
4. 对不可逆矩阵求逆导致报错。
5. 能用 `np.linalg.solve` 时却手动求逆，数值稳定性更差。
