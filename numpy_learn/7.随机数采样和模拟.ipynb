## 1. 随机数生成器

推荐使用 `np.random.default_rng()` 创建随机数生成器。它比旧的 `np.random.seed()` 风格更清晰。


```python
import numpy as np

# 固定种子后，每次运行都会得到相同的随机结果，方便学习和复现实验。
rng = np.random.default_rng(seed=42)

# 生成 0 到 1 之间的随机小数。
print(rng.random(5))

# 生成指定形状的随机小数数组。
print(rng.random((2, 3)))

# 生成随机整数，low 包含，high 不包含。
print(rng.integers(low=1, high=10, size=5))
```

### 解释

- `seed` 是随机种子，用于保证结果可复现。
- `random(size)` 生成 `[0, 1)` 之间的随机小数。
- `integers(low, high, size)` 生成整数，`high` 取不到。


## 2. 常用随机方法表

| 方法 | 作用 |
| --- | --- |
| `rng.random(size)` | 生成 0 到 1 之间的随机小数 |
| `rng.integers(low, high, size)` | 生成随机整数 |
| `rng.normal(loc, scale, size)` | 生成正态分布随机数 |
| `rng.uniform(low, high, size)` | 生成均匀分布随机数 |
| `rng.choice(data, size, replace, p)` | 随机抽样 |
| `rng.shuffle(arr)` | 原地打乱数组 |
| `rng.permutation(arr)` | 返回打乱后的新数组 |


```python
import numpy as np

rng = np.random.default_rng(0)

# 均匀分布：每个区间出现概率大致相同。
uniform_data = rng.uniform(low=10, high=20, size=5)
print('均匀分布：', uniform_data)

# 正态分布：loc 是均值，scale 是标准差。
normal_data = rng.normal(loc=60, scale=10, size=10)
print('正态分布：', normal_data)

# choice 从给定数据中抽样。
names = np.array(['Tom', 'Jerry', 'Alice', 'Bob'])
print('随机抽 2 个：', rng.choice(names, size=2, replace=False))

# 带概率抽样。
prizes = np.array(['一等奖', '二等奖', '三等奖'])
prob = np.array([0.1, 0.3, 0.6])
print('按概率抽奖：', rng.choice(prizes, size=5, replace=True, p=prob))

# permutation 返回新数组，不修改原数组。
arr = np.arange(1, 6)
print('打乱后的新数组：', rng.permutation(arr))
print('原数组：', arr)

# shuffle 原地打乱，会修改原数组。
rng.shuffle(arr)
print('shuffle 后原数组：', arr)
```

### 解释

- `replace=False` 表示不放回抽样，抽过的元素不会再次出现。
- `p` 表示每个元素被抽中的概率，所有概率之和必须为 1。
- `permutation` 不修改原数组。
- `shuffle` 会直接修改原数组。


## 3. 随机模拟案例

随机数常用于模拟实验，例如掷骰子、抽奖、蒙特卡洛估计。


```python
import numpy as np

rng = np.random.default_rng(123)

# 模拟掷骰子 10000 次。
dice = rng.integers(1, 7, size=10000)

# 统计每个点数出现次数。
values, counts = np.unique(dice, return_counts=True)
print(values)
print(counts)
print('每个点数出现比例：', counts / counts.sum())

# 模拟考试成绩，假设平均分 75，标准差 10。
scores = rng.normal(loc=75, scale=10, size=1000)

# 把成绩限制在 0 到 100 之间。
scores = np.clip(scores, 0, 100)

print('平均分：', scores.mean())
print('最高分：', scores.max())
print('最低分：', scores.min())
print('及格率：', np.mean(scores >= 60))
```

### 解释

- `integers(1, 7)` 可以模拟骰子，因为 7 取不到。
- `unique(..., return_counts=True)` 可以统计每个点数出现次数。
- `np.clip(scores, 0, 100)` 把小于 0 的值变成 0，大于 100 的值变成 100。
- 布尔数组可以直接求平均，`True` 当作 1，`False` 当作 0。


## 4. 本章练习题

1. 固定随机种子 2026，生成 10 个 1 到 100 之间的随机整数。
2. 从姓名数组中不放回抽取 3 个名字。
3. 模拟抛硬币 1000 次，统计正面比例。
4. 生成 100 个正态分布成绩，限制在 0 到 100 之间。
5. 打乱数组 `[1, 2, 3, 4, 5]`，保留原数组不变。


```python
import numpy as np

rng = np.random.default_rng(2026)

# 1. 生成随机整数
nums = rng.integers(1, 101, size=10)
print(nums)

# 2. 不放回抽样
names = np.array(['Tom', 'Jerry', 'Alice', 'Bob', 'Lucy'])
print(rng.choice(names, size=3, replace=False))

# 3. 模拟抛硬币，1 表示正面，0 表示反面
coins = rng.integers(0, 2, size=1000)
print('正面比例：', coins.mean())

# 4. 生成成绩并限制范围
scores = rng.normal(75, 12, size=100)
scores = np.clip(scores, 0, 100)
print(scores[:10])
print(scores.mean())

# 5. permutation 不修改原数组
arr = np.array([1, 2, 3, 4, 5])
shuffled = rng.permutation(arr)
print(shuffled)
print(arr)
```

## 5. 常见错误总结

1. 旧式 `np.random.seed()` 和新版 `default_rng()` 混用。
2. 忘记 `integers` 的 high 取不到。
3. 不放回抽样时，抽样数量超过总体数量。
4. 概率 `p` 的总和不等于 1。
5. 分不清 `shuffle` 会修改原数组，`permutation` 不会。
