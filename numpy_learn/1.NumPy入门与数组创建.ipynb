## 1. 为什么需要 NumPy

Python 列表可以保存数据，但它不擅长大规模数值计算。NumPy 数组支持向量化运算，可以一次性对整组数据做计算，代码更短，速度也通常更快。


```python
import numpy as np

# Python 列表做整体加法时，不是每个元素加 10，而是列表拼接。
python_list = [1, 2, 3, 4]
print('Python 列表 + 列表：', python_list + [10])

# NumPy 数组支持向量化运算。
# arr + 10 表示数组中的每个元素都加 10。
arr = np.array([1, 2, 3, 4])
print('NumPy 数组 + 10：', arr + 10)

# 数组之间也可以逐元素计算。
arr2 = np.array([10, 20, 30, 40])
print('两个数组相加：', arr + arr2)
print('两个数组相乘：', arr * arr2)
```

### 解释

- `np.array(...)` 可以把列表转换成 NumPy 数组。
- `arr + 10` 会把 10 加到数组的每一个元素上。
- `arr + arr2` 会按相同位置逐元素相加。
- 这种不需要手写循环、直接对整组数据运算的方式叫向量化。


## 2. 创建数组：`np.array`

`np.array()` 是最基础的数组创建方式，可以从列表、元组、嵌套列表创建数组。


```python
import numpy as np

# 一维数组：可以理解成一行数据。
a = np.array([1, 2, 3, 4])
print(a)
print(type(a))

# 二维数组：可以理解成表格或矩阵。
b = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(b)

# 创建数组时可以指定 dtype，也就是数组元素的数据类型。
c = np.array([1, 2, 3], dtype=float)
print(c)
print(c.dtype)

# 注意：NumPy 数组通常要求元素类型统一。
# 如果混合整数和浮点数，NumPy 会自动转换成更通用的浮点类型。
d = np.array([1, 2.5, 3])
print(d)
print(d.dtype)
```

### 解释

- `ndarray` 是 NumPy 数组的类型。
- 二维数组要求每一行长度一致，否则可能得到对象数组或报错。
- `dtype` 表示数据类型，例如 `int64`、`float64`。
- 数值计算时，统一类型让底层计算更高效。


## 3. 常用数组创建方法表

| 方法 | 作用 | 示例 |
| --- | --- | --- |
| `np.array(data)` | 从列表等数据创建数组 | `np.array([1, 2, 3])` |
| `np.arange(start, stop, step)` | 创建等差整数/浮点序列，不包含 stop | `np.arange(1, 10, 2)` |
| `np.linspace(start, stop, num)` | 创建指定数量的等间隔数据，默认包含 stop | `np.linspace(0, 1, 5)` |
| `np.zeros(shape)` | 创建全 0 数组 | `np.zeros((2, 3))` |
| `np.ones(shape)` | 创建全 1 数组 | `np.ones((3, 2))` |
| `np.full(shape, value)` | 创建指定填充值的数组 | `np.full((2, 2), 7)` |
| `np.eye(n)` | 创建单位矩阵 | `np.eye(3)` |
| `np.empty(shape)` | 创建未初始化数组 | `np.empty((2, 2))` |
| `np.zeros_like(arr)` | 创建形状相同的全 0 数组 | `np.zeros_like(arr)` |
| `np.ones_like(arr)` | 创建形状相同的全 1 数组 | `np.ones_like(arr)` |


```python
import numpy as np

# arange 类似 Python 的 range，但返回 NumPy 数组。
# 结束值 10 取不到。
print(np.arange(1, 10, 2))

# linspace 常用于生成坐标、刻度、采样点。
# 0 到 1 之间生成 5 个等间隔点。
print(np.linspace(0, 1, 5))

# zeros 创建全 0 数组，shape=(2, 3) 表示 2 行 3 列。
print(np.zeros((2, 3)))

# ones 创建全 1 数组。
print(np.ones((3, 2)))

# full 创建全部填充为指定值的数组。
print(np.full((2, 2), 7))

# eye 创建单位矩阵：主对角线是 1，其余位置是 0。
print(np.eye(3))

# empty 创建未初始化数组。
# 它只申请内存，不保证里面的值是什么，所以初学阶段不建议用它保存正式数据。
print(np.empty((2, 2)))

# like 系列会参考已有数组的形状。
base = np.array([[1, 2, 3], [4, 5, 6]])
print(np.zeros_like(base))
print(np.ones_like(base))
```

### 解释

- `arange` 更像按步长生成，适合整数序列。
- `linspace` 更像按数量生成，适合浮点采样。
- `zeros`、`ones`、`full` 常用于初始化数组。
- `eye` 常用于线性代数中的单位矩阵。
- `empty` 速度快，但内容不确定，必须后续手动赋值。


## 4. 本章练习题

1. 创建数组 `[10, 20, 30, 40]`，让每个元素加 5。
2. 使用 `np.arange()` 创建 1 到 20 中的所有偶数。
3. 使用 `np.linspace()` 创建 0 到 100 之间的 6 个等间隔数字。
4. 创建一个 3 行 4 列的全 0 数组。
5. 创建一个 4 阶单位矩阵。


```python
import numpy as np

# 1. 每个元素加 5
arr = np.array([10, 20, 30, 40])
print(arr + 5)

# 2. 1 到 20 中的所有偶数
# stop 写 21，是因为 arange 的结束值取不到。
even_numbers = np.arange(2, 21, 2)
print(even_numbers)

# 3. 0 到 100 之间生成 6 个等间隔数字
points = np.linspace(0, 100, 6)
print(points)

# 4. 3 行 4 列的全 0 数组
zeros = np.zeros((3, 4))
print(zeros)

# 5. 4 阶单位矩阵
identity = np.eye(4)
print(identity)
```

## 5. 常见错误总结

1. 把 Python 列表和 NumPy 数组的加法混淆。
2. `np.arange()` 的结束值取不到。
3. 二维数组每一行长度不一致。
4. 忘记导入 `import numpy as np`。
5. 误以为 `np.empty()` 会自动填 0。
