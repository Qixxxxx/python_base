## 1. 基础数学运算

NumPy 数组支持逐元素加减乘除、幂运算、取余等操作。


```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print(a + b)
print(b - a)
print(a * b)
print(b / a)
print(a ** 2)
print(b % 3)

# 数组和标量运算时，标量会作用到每个元素上。
print(a + 100)
print(a * 10)
```

### 解释

- 数组形状相同，运算会逐元素进行。
- 数组和标量运算时，标量会广播到每个元素。
- 除法 `/` 结果通常是浮点数。


## 2. 通用函数 ufunc 表

通用函数是 NumPy 中对数组逐元素计算的函数。

| 函数 | 作用 |
| --- | --- |
| `np.abs()` | 绝对值 |
| `np.sqrt()` | 平方根 |
| `np.exp()` | 指数函数 |
| `np.log()` | 自然对数 |
| `np.sin()`、`np.cos()` | 三角函数 |
| `np.round()` | 四舍五入 |
| `np.floor()` | 向下取整 |
| `np.ceil()` | 向上取整 |
| `np.maximum(a, b)` | 逐元素取较大值 |
| `np.minimum(a, b)` | 逐元素取较小值 |


```python
import numpy as np

arr = np.array([-4, -1, 0, 1, 4])

print('绝对值：', np.abs(arr))
print('平方根：', np.sqrt(np.array([1, 4, 9, 16])))
print('指数：', np.exp(np.array([0, 1, 2])))
print('自然对数：', np.log(np.array([1, np.e, np.e ** 2])))

decimal = np.array([1.2, 1.5, 1.8, -1.2])
print('round：', np.round(decimal))
print('floor：', np.floor(decimal))
print('ceil：', np.ceil(decimal))

a = np.array([1, 5, 3])
b = np.array([2, 4, 6])
print('逐元素最大值：', np.maximum(a, b))
print('逐元素最小值：', np.minimum(a, b))
```

### 解释

- ufunc 不需要手写循环，会对数组中每个元素执行同一种计算。
- `maximum` 和 Python 内置 `max` 不同，它是逐元素比较。
- 对数函数要求输入为正数，否则会产生警告或 NaN。


## 3. 统计函数和 `axis`

统计函数可以对整个数组计算，也可以沿指定轴计算。


```python
import numpy as np

scores = np.array([
    [86, 78, 92],
    [91, 88, 95],
    [79, 90, 84],
    [88, 76, 81]
])

print('总和：', np.sum(scores))
print('平均分：', np.mean(scores))
print('最大值：', np.max(scores))
print('最小值：', np.min(scores))
print('标准差：', np.std(scores))

# axis=0 表示沿着行的方向压缩，结果按列统计。
print('每一科平均分：', np.mean(scores, axis=0))

# axis=1 表示沿着列的方向压缩，结果按行统计。
print('每个学生平均分：', np.mean(scores, axis=1))

# argmax 返回最大值所在的位置索引。
print('所有成绩最高值位置：', np.argmax(scores))

# unravel_index 可以把扁平索引转换成二维坐标。
row, col = np.unravel_index(np.argmax(scores), scores.shape)
print('最高分位置：', row, col)
```

### 解释

- 不写 `axis` 时，对所有元素统计。
- `axis=0` 常理解为“按列统计”。
- `axis=1` 常理解为“按行统计”。
- `argmax`、`argmin` 返回的是位置，不是最大/最小值本身。


## 4. 累计函数

累计函数适合处理累计和、累计乘积、累计最大值等问题。


```python
import numpy as np

arr = np.array([1, 2, 3, 4])

print('累计和：', np.cumsum(arr))
print('累计乘积：', np.cumprod(arr))

sales = np.array([100, 120, 90, 150, 130])
print('销售额累计：', np.cumsum(sales))

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print('按列累计和：')
print(np.cumsum(matrix, axis=0))

print('按行累计和：')
print(np.cumsum(matrix, axis=1))
```

### 解释

- `cumsum` 返回每个位置之前所有元素的累计和。
- `cumprod` 返回累计乘积。
- 指定 `axis` 后，可以按行或按列累计。


## 5. 本章练习题

1. 创建数组 `[1, 4, 9, 16]`，计算平方根。
2. 创建成绩矩阵，计算每个学生平均分。
3. 计算每门课最高分。
4. 找出整个矩阵中最低分的位置。
5. 计算销售额数组的累计销售额。


```python
import numpy as np

# 1. 平方根
arr = np.array([1, 4, 9, 16])
print(np.sqrt(arr))

# 2. 每个学生平均分
scores = np.array([
    [86, 78, 92],
    [91, 88, 95],
    [79, 90, 84]
])
print(np.mean(scores, axis=1))

# 3. 每门课最高分
print(np.max(scores, axis=0))

# 4. 最低分位置
row, col = np.unravel_index(np.argmin(scores), scores.shape)
print(row, col, scores[row, col])

# 5. 累计销售额
sales = np.array([100, 150, 120, 180])
print(np.cumsum(sales))
```

## 6. 常见错误总结

1. 不理解 `axis=0` 和 `axis=1` 的方向。
2. 把 `np.maximum` 和内置 `max` 混淆。
3. 对负数开平方得到 NaN。
4. 忘记 `argmax` 返回的是索引。
5. 统计包含 NaN 的数组时没有使用 `nanmean` 等函数。
