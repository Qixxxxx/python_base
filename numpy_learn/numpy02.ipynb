```python
# 指定轴比较大小，axis=0表示行与行之间比较大小
a.max(axis=0), a.min(axis=0), a.sum(axis=0)
```

```python
a.max(axis=1), a.min(axis=1), a.sum(axis=1)
```

```python
a.mean(), a.std(), a.var()  # 均值， 标准差， 方差
```

```python
a.mean(axis=0), a.std(axis=0), a.var(axis=0),a.mean(axis=1), a.std(axis=1), a.var(axis=1)
```

```python
a.argmax(), a.argmin()  # 最大值的索引,最小值的索引
```

```python
a.argmax(axis=0), a.argmin(axis=0), a.argmax(axis=1), a.argmin(axis=1)
```

## 数组形状操作

```python
a = np.arange(12).reshape(3, 4)
a, a.reshape(4, 3)
```

```python
"""
reshape和resize是有区别的，reshape不会改变原数组会输出改变后的数组，resize会直接改变原数组无输出值
"""
a.resize(4, 3)
a
```

```python
"""
展平操作：flatten和ravel：由ravel创建的数组会影响原数组，而flatten不会
"""
a = np.arange(12).reshape(3, 4)
b = a.flatten()
c = a.ravel()
a,b,c
```

```python
b[3] = 100
a, b
```

```python
c[3] = 100   # 可以看到a也随之发生改变
a, c
```

## 数组堆叠

```python
a = np.array([[9., 7.],
              [5., 2.]])
b = np.array([[1., 9.],
              [5., 1.]])
c = np.vstack((a, b))   # 将行进行堆叠
d = np.hstack((a, b))   # 将列进行堆叠
c, d
```

```python
"""
stack是在维度上添加新的维度，和vstack hstack不一致
"""
e = np.stack((a, b), axis=0)
f = np.stack((a, b), axis=1) 
e, e.shape, f, f.shape
```

```python
# concat和hstack vstack一致
np.concatenate((a, b), axis=0), np.concatenate((a, b), axis=1)
```

## 数组切分

```python
a = np.array([[6., 7., 6., 9., 0., 5., 4., 0., 6., 8., 5., 2.],
              [8., 5., 5., 7., 1., 8., 6., 7., 1., 8., 1., 0.]])
```

```python
np.hsplit(a,2)      # 将a对列切2份      
```

```python
np.vsplit(a,2)      # 将a对行切2份
```

```python
np.split(a,2)      # 将a对行切2份
```

```python
np.split(a, 3, axis=1)      # 将a对列切3份
```

```python
# 实现不均匀切分
a = np.hsplit(a, (3, 4))      # 在列索引为3处切，在列索引为4处切
a ,a[1]
```
