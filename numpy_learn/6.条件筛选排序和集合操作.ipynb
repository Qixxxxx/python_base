## 1. 布尔索引

比较运算会返回布尔数组，布尔数组可以用于筛选元素。


```python
import numpy as np

scores = np.array([55, 80, 92, 40, 67, 100])

# 比较运算返回布尔数组。
mask = scores >= 60
print(mask)

# 使用布尔数组筛选及格成绩。
passed_scores = scores[mask]
print(passed_scores)

# 可以直接把条件写进中括号。
print(scores[scores < 60])
```

### 解释

- `scores >= 60` 会对每个元素判断是否成立。
- 布尔索引只保留对应位置为 `True` 的元素。
- 布尔数组长度必须和被筛选的维度长度一致。


## 2. 多条件筛选

NumPy 中多个条件组合不能直接写 `and`、`or`，要使用 `&` 和 `|`，并给每个条件加括号。


```python
import numpy as np

scores = np.array([55, 80, 92, 40, 67, 100])

# 筛选 60 到 90 之间的成绩。
middle = scores[(scores >= 60) & (scores <= 90)]
print(middle)

# 筛选小于 60 或大于等于 90 的成绩。
special = scores[(scores < 60) | (scores >= 90)]
print(special)

# 取反使用 ~。
not_failed = scores[~(scores < 60)]
print(not_failed)
```

### 解释

- `&` 表示并且。
- `|` 表示或者。
- `~` 表示取反。
- 每个条件必须用括号包起来，否则可能因为运算优先级导致错误。


## 3. `np.where`

`np.where` 可以根据条件生成新数组，也可以返回满足条件的位置。


```python
import numpy as np

scores = np.array([55, 80, 92, 40, 67, 100])

# 三参数形式：条件成立取第 2 个参数，否则取第 3 个参数。
labels = np.where(scores >= 60, '及格', '不及格')
print(labels)

# 把不及格成绩替换成 0。
new_scores = np.where(scores < 60, 0, scores)
print(new_scores)

# 单参数形式：返回满足条件的位置索引。
index = np.where(scores >= 90)
print(index)
print(scores[index])
```

### 解释

- `np.where(condition, x, y)` 类似批量版 `x if condition else y`。
- 单参数 `np.where(condition)` 返回满足条件的位置。
- 如果只是筛选元素，布尔索引通常更直接。


## 4. 排序方法表

| 方法 | 作用 |
| --- | --- |
| `np.sort(arr)` | 返回排序后的新数组 |
| `arr.sort()` | 原地排序，修改原数组 |
| `np.argsort(arr)` | 返回排序后的索引顺序 |
| `np.lexsort(keys)` | 按多个键排序 |
| `np.argmax(arr)` | 最大值位置 |
| `np.argmin(arr)` | 最小值位置 |


```python
import numpy as np

scores = np.array([80, 95, 88, 76])
names = np.array(['Tom', 'Jerry', 'Alice', 'Bob'])

print('排序后的成绩：', np.sort(scores))

# argsort 返回从小到大排序时的索引顺序。
order = np.argsort(scores)
print('排序索引：', order)
print('按成绩从小到大排列姓名：', names[order])
print('按成绩从小到大排列成绩：', scores[order])

# 从大到小可以把索引反转。
desc_order = order[::-1]
print('按成绩从大到小排列姓名：', names[desc_order])
print('按成绩从大到小排列成绩：', scores[desc_order])

print('最高分学生：', names[np.argmax(scores)])
print('最低分学生：', names[np.argmin(scores)])
```

### 解释

- `np.sort` 不修改原数组。
- `arr.sort()` 会直接修改原数组。
- `argsort` 在需要让多个数组同步排序时非常有用。
- `names[order]` 表示按成绩排序得到的索引重排姓名。


## 5. 去重和集合操作

NumPy 提供了类似集合的数组操作，适合处理一维数组。


```python
import numpy as np

a = np.array([1, 2, 2, 3, 4, 4, 5])
b = np.array([4, 5, 6, 7])

print('去重：', np.unique(a))
print('交集：', np.intersect1d(a, b))
print('并集：', np.union1d(a, b))
print('差集 a-b：', np.setdiff1d(a, b))
print('是否在 b 中：', np.isin(a, b))
print('a 中属于 b 的元素：', a[np.isin(a, b)])

# unique 也可以返回出现次数。
values, counts = np.unique(a, return_counts=True)
print(values)
print(counts)
```

### 解释

- `unique` 默认会排序。
- `intersect1d` 求交集。
- `union1d` 求并集。
- `setdiff1d(a, b)` 求在 a 中但不在 b 中的元素。
- `isin` 返回布尔数组，常用于筛选。


## 6. 本章练习题

1. 筛选出数组中大于 50 的元素。
2. 筛选出 60 到 90 之间的成绩。
3. 使用 `np.where` 把低于 60 的成绩标记为“不及格”。
4. 根据成绩对姓名从高到低排序。
5. 统计数组中每个数字出现的次数。


```python
import numpy as np

# 1. 大于 50
arr = np.array([10, 60, 30, 80, 50])
print(arr[arr > 50])

# 2. 60 到 90 之间
scores = np.array([55, 80, 92, 40, 67, 100])
print(scores[(scores >= 60) & (scores <= 90)])

# 3. 标记及格状态
print(np.where(scores < 60, '不及格', '及格'))

# 4. 按成绩从高到低排序姓名
names = np.array(['Tom', 'Jerry', 'Alice', 'Bob'])
score2 = np.array([80, 95, 88, 76])
order = np.argsort(score2)[::-1]
print(names[order])
print(score2[order])

# 5. 统计出现次数
nums = np.array([1, 2, 2, 3, 3, 3])
values, counts = np.unique(nums, return_counts=True)
print(values)
print(counts)
```

## 7. 常见错误总结

1. 多条件筛选时使用 `and`、`or`，应该使用 `&`、`|`。
2. 多条件忘记加括号。
3. 误以为 `np.sort` 会修改原数组。
4. `argsort` 返回的是索引，不是排序后的值。
5. `unique` 默认会排序，原始顺序可能改变。
