```python
import numpy as np
```

## 随机数

```python
# np.random.rand(d0,d1,...,dn) 可以返回一组n维度服从“0~1”均匀分布的随机样本值。随机样本取值范围是[0,1)
np.random.rand(5)
```

```python
np.random.rand(2,3) # 生成二维数组
```

```python
np.random.rand()    # 生成一个标量
```

```python
# np.random.normal(loc=0,scale=1e-2,size=shape) 可以返回一组n维度服从正态分布的随机样本值。
# 参数loc(float)：正态分布的均值，对应着这个分布的中心。loc=0说明这一个以Y轴为对称轴的正态分布
# 参数scale(float)：正态分布的标准差，对应分布的宽度，scale越大，正态分布的曲线越矮胖，scale越小，曲线越高瘦。
# 参数size(int 或者整数元组)：输出的值赋在shape里，默认为None
np.random.normal(0, 0.1, 10)
```

```python
# np.random.randn(d0,d1,...,dn) 可以返回一组n维度服从标准正态分布的随机样本值。
np.random.randn(2,5)
```

```python
# np.random.randint(low,high,size,)返回low<= n <high范围的整数
np.random.randint(0, 4, size=10)  # 单值则为最大值
```

```python
np.random.randint(2, size=(2,3)) # size可以生成指定形状数组
```
